```python
from google.colab import files
import os

# 1. Upload do notebook
uploaded = files.upload()

# 2. Localizar automaticamente o arquivo .ipynb enviado
notebooks = [
    nome for nome in uploaded.keys()
    if nome.lower().endswith(".ipynb")
]

if not notebooks:
    raise ValueError(
        "Nenhum arquivo .ipynb foi enviado."
    )

if len(notebooks) > 1:
    raise ValueError(
        "Envie apenas um arquivo .ipynb por vez."
    )

notebook = notebooks[0]

print(f"✓ Notebook identificado: {notebook}")

# 3. Nome fixo do relatório HTML
html_output = "technical-analysis.html"

# 4. Gerar HTML
!jupyter nbconvert \
    --to html \
    "{notebook}" \
    --output "{html_output}"

# 5. Verificar e baixar
if os.path.exists(html_output):
    print(f"✓ HTML gerado: {html_output}")
    files.download(html_output)
else:
    raise FileNotFoundError(
        f"O arquivo {html_output} não foi gerado."
    )
```
