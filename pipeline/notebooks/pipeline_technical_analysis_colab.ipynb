## 2. Analysis Criteria

A análise foi conduzida a partir dos artefatos técnicos versionados associados às entregas V1, V2, V3 e V4, considerando tanto a presença das práticas quanto o nível de evidência disponível para sustentá-las. Para preservar a rastreabilidade entre os artefatos observados e as conclusões apresentadas, foram estabelecidos critérios para a classificação das evidências e para a avaliação da maturidade das práticas de automação do pipeline.

###2.1 Evidence Classification

As evidências foram analisadas considerando três níveis distintos:

Configuração: a capacidade está declarada em um artefato técnico versionado, indicando que foi configurada no projeto;

Orquestração no pipeline: a capacidade é explicitamente invocada ou incorporada à definição do pipeline de CI;

Execução efetiva: além da configuração ou orquestração, existem evidências de runtime que comprovam sua execução, como histórico de jobs, logs, branches ou Merge Requests gerados pela ferramenta.

Essa distinção evita inferir a execução de uma prática exclusivamente a partir da existência de sua configuração. Dessa forma, quando os artefatos analisados demonstram a configuração de uma capacidade, mas não fornecem evidências suficientes de sua execução efetiva, essa condição é explicitamente indicada na análise. Essa abordagem é particularmente relevante para a análise do Renovate, cuja configuração e orquestração podem ser verificadas nos artefatos versionados, enquanto sua execução efetiva depende de evidências adicionais do ambiente de CI.

Para representação das evidências nas matrizes, são utilizados os seguintes símbolos:

- **✓** para capacidade evidenciada ou configurada no artefato analisado;
- **—** para capacidade ausente ou não evidenciada no artefato analisado;
- **?** quando a confirmação depende de evidência adicional de execução.

A ausência de evidência de execução não é interpretada automaticamente como ausência da prática, preservando a distinção entre o que pode ser diretamente observado nos artefatos e aquilo que depende de evidências complementares.

###2.2 Pipeline Maturity Classification

Além da classificação das evidências individuais, foi adotada uma escala ordinal de quatro níveis para sintetizar o estágio de maturidade das práticas de automação do pipeline no recorte analisado.

Nível	Classificação	Critério:

1	`Ausente/Tradicional`	Práticas inexistentes ou totalmente manuais.

Ausência de pipeline versionado ou pipeline sem etapas relevantes de automação; build, testes, análise de qualidade e entrega predominantemente manuais.

2	`Inicial/Ad hoc`	Adoção pontual e não sistematizada.

Pipeline existente, mas com automações pontuais, por exemplo apenas build, análise estática ou alguma tarefa isolada; baixa integração entre etapas.

3	`Parcial/Emergente`	Práticas estruturadas, porém incompletas.

Pipeline versionado com práticas estruturadas de CI, como restore, build, análise de qualidade, gerenciamento automatizado de dependências ou outras etapas recorrentes, mas ainda com lacunas relevantes, como ausência de testes automatizados, deploy, promoção ou rollback.

4	`Avançado/Consolidado`	Práticas institucionalizadas e integradas.

Pipeline integrado cobrindo build, testes automatizados, controles de qualidade efetivamente bloqueantes e mecanismos estruturados de entrega/implantação, promoção e recuperação. A simples presença dos arquivos não basta: são necessárias também evidências de execução e continuidade dessas práticas.

A classificação é realizada a partir do conjunto das evidências observadas nas versões analisadas, considerando a existência e a continuidade das práticas de Integração Contínua, automação de build, análise de qualidade, gerenciamento de dependências, testes automatizados e mecanismos relacionados à entrega e implantação do software.

A escala é aplicada especificamente às práticas de automação do pipeline observadas no recorte temporal analisado. Portanto, o nível resultante não deve ser interpretado isoladamente como uma classificação da maturidade DevOps global do projeto ou da organização. Uma avaliação global exigiria a aplicação de critérios equivalentes às demais dimensões envolvidas na adoção de DevOps.

A classificação final do pipeline é apresentada somente após a análise individual das versões e a análise longitudinal, permitindo que o nível atribuído seja derivado das evidências técnicas apresentadas ao longo do relatório, e não estabelecido previamente como resultado da avaliação.

3. AI-Assisted Interpretation

A Inteligência Artificial é utilizada neste notebook como mecanismo de apoio à interpretação das evidências técnicas extraídas dos artefatos analisados.

A IA não realiza a identificação primária das práticas nem substitui os critérios objetivos definidos para a classificação da maturidade do pipeline. As evidências são previamente extraídas de forma automatizada a partir dos arquivos fornecidos, estruturadas e avaliadas segundo uma rubrica previamente estabelecida.

A partir dessas informações, a IA é utilizada para:

- interpretar o conjunto de evidências identificadas;
- verificar se as evidências sustentam a classificação obtida pelos critérios definidos;
- elaborar uma justificativa técnica e acadêmica para a classificação;
- identificar lacunas e limitações da análise;
- distinguir evidências de configuração, orquestração e execução efetiva;
- evitar inferências sobre práticas que não estejam diretamente sustentadas pelos artefatos analisados.

A interpretação assistida por IA deve ser conservadora e rastreável. A ausência de evidência não deve ser automaticamente interpretada como inexistência da prática, especialmente quando o artefato necessário para sua verificação não estiver disponível.

Assim, o fluxo analítico adotado neste notebook segue a sequência:

** `1`artefatos técnicos → `2`identificação e validação dos arquivos → `3`extração objetiva de evidências → `4`classificação determinística por critérios → `5`estruturação das evidências → `6`interpretação assistida por IA → `7`armazenamento dos resultados por versão → `8`consolidação das evidências → `9`análise longitudinal**

A classificação básica permanece reproduzível independentemente da IA, enquanto o modelo é utilizado como apoio à análise qualitativa e à elaboração da interpretação dos resultados.

```python
# Common Utility Functions
def contem(texto, *padroes):
    if texto is None:
        return False

    texto = str(texto).lower()

    return any(
        str(padrao).lower() in texto
        for padrao in padroes
    )


def buscar_regex(texto, padrao):
    if texto is None:
        return None

    return re.search(
        padrao,
        str(texto),
        flags=re.IGNORECASE | re.MULTILINE
    )
```

```python
# ---------------------------------------------------------
# AUTOMATED EVIDENCE EXTRACTION
# ---------------------------------------------------------

target_framework = None
sdk_pipeline = None

# Reinicia as evidências a cada execução
evidencias = {}


# ---------------------------------------------------------
# 1. .csproj
# ---------------------------------------------------------

if csproj:

    try:
        root = ET.fromstring(csproj)

        for elemento in root.iter():

            tag = elemento.tag.split("}")[-1]

            if tag == "TargetFramework":

                target_framework = (
                    elemento.text.strip()
                    if elemento.text
                    else None
                )

                break

            elif tag == "TargetFrameworks":

                target_framework = (
                    elemento.text.strip()
                    if elemento.text
                    else None
                )

                break

    except ET.ParseError as erro:

        print(
            "⚠ Não foi possível interpretar o arquivo .csproj:"
        )

        print(erro)


# ---------------------------------------------------------
# 2. GitLab CI
# ---------------------------------------------------------

if pipeline:

    # SDK .NET utilizado no pipeline
    match_sdk = buscar_regex(
        pipeline,
        r"mcr\.microsoft\.com/dotnet/sdk:([0-9.]+)"
    )

    if match_sdk:
        sdk_pipeline = match_sdk.group(1)


    # Pipeline versionado
    evidencias["GitLab CI como código"] = True


    # Restore
    evidencias["`dotnet restore` no CI"] = contem(
        pipeline,
        "dotnet restore"
    )


    # Build Release
    possui_build = contem(
        pipeline,
        "dotnet build"
    )

    possui_release = contem(
        pipeline,
        "--configuration release",
        "-c release"
    )

    evidencias["Build Release no CI"] = (
        possui_build
        and possui_release
    )


    # SonarQube
    evidencias["SonarQube"] = contem(
        pipeline,
        "dotnet-sonarscanner",
        "sonarscanner",
        "sonar.host.url",
        "sonar_token"
    )


    # Quality Gate
    evidencias["Espera pelo Quality Gate"] = contem(
        pipeline,
        "sonar.qualitygate.wait=true"
    )


    # Quality Gate bloqueante
    quality_gate_wait = evidencias[
        "Espera pelo Quality Gate"
    ]

    allow_failure_match = buscar_regex(
        pipeline,
        r"allow_failure\s*:\s*true"
    )

    allow_failure_true = (
        allow_failure_match is not None
    )

    evidencias["Quality Gate bloqueante"] = (
        quality_gate_wait
        and not allow_failure_true
    )


    # Cache Sonar
    evidencias["Cache Sonar"] = contem(
        pipeline,
        "${sonar_user_home}/cache",
        ".sonar"
    )


    # Cache NuGet
    evidencias["Cache NuGet"] = contem(
        pipeline,
        ".nuget/packages"
    )


    # Merge Request
    evidencias["Regra para Merge Request"] = contem(
        pipeline,
        "merge_request_event"
    )


    # Branch main
    possui_variavel_branch = contem(
        pipeline,
        "ci_commit_branch"
    )

    possui_main_match = buscar_regex(
        pipeline,
        r"ci_commit_branch.*main"
    )

    possui_main = (
        possui_main_match is not None
    )

    evidencias["Regra para branch main"] = (
        possui_variavel_branch
        and possui_main
    )


    # Testes automatizados
    evidencias["Testes automatizados"] = contem(
        pipeline,
        "dotnet test",
        "pytest",
        "npm test",
        "npm run test",
        "mvn test",
        "gradle test"
    )


    # Deploy automatizado
    evidencias["Deploy automatizado"] = contem(
        pipeline,
        "kubectl apply",
        "helm upgrade",
        "docker stack deploy",
        "docker compose up",
        "docker-compose up",
        "oc apply"
    )


    # Promoção automática
    evidencias["Promoção automática"] = contem(
        pipeline,
        "promote",
        "promotion",
        "promote-to-production",
        "promote_to_production"
    )


    # Rollback automatizado
    evidencias["Rollback automatizado"] = contem(
        pipeline,
        "helm rollback",
        "kubectl rollout undo",
        "docker service rollback"
    )

else:

    evidencias["GitLab CI como código"] = None
    evidencias["`dotnet restore` no CI"] = None
    evidencias["Build Release no CI"] = None
    evidencias["SonarQube"] = None
    evidencias["Espera pelo Quality Gate"] = None
    evidencias["Quality Gate bloqueante"] = None
    evidencias["Cache Sonar"] = None
    evidencias["Cache NuGet"] = None
    evidencias["Regra para Merge Request"] = None
    evidencias["Regra para branch main"] = None
    evidencias["Testes automatizados"] = None
    evidencias["Deploy automatizado"] = None
    evidencias["Promoção automática"] = None
    evidencias["Rollback automatizado"] = None


# ---------------------------------------------------------
# 3. Renovate
# ---------------------------------------------------------

if renovate_json:

    evidencias["Renovate configurado"] = True

    try:

        renovate_data = json.loads(
            renovate_json
        )

        enabled_managers = renovate_data.get(
            "enabledManagers",
            []
        )

        evidencias[
            "Gerenciamento NuGet pelo Renovate"
        ] = (
            "nuget" in enabled_managers
        )

        evidencias[
            "Gerenciamento Dockerfile pelo Renovate"
        ] = (
            "dockerfile" in enabled_managers
        )


        # Vulnerability Alerts
        vulnerability_alerts = renovate_data.get(
            "vulnerabilityAlerts"
        )

        if vulnerability_alerts is not None:

            evidencias[
                "Vulnerability Alerts"
            ] = True

        else:

            evidencias[
                "Vulnerability Alerts"
            ] = contem(
                renovate_json,
                "vulnerabilityalerts",
                "osv"
            )


        # Lock File Maintenance
        lock_file_maintenance = renovate_data.get(
            "lockFileMaintenance"
        )

        if lock_file_maintenance is not None:

            evidencias[
                "Lock File Maintenance"
            ] = True

        else:

            evidencias[
                "Lock File Maintenance"
            ] = contem(
                renovate_json,
                "lockfilemaintenance"
            )


    except json.JSONDecodeError:

        print(
            "⚠ renovate.json inválido. "
            "A análise será feita por busca textual."
        )

        evidencias[
            "Gerenciamento NuGet pelo Renovate"
        ] = contem(
            renovate_json,
            '"nuget"'
        )

        evidencias[
            "Gerenciamento Dockerfile pelo Renovate"
        ] = contem(
            renovate_json,
            '"dockerfile"'
        )

        evidencias[
            "Vulnerability Alerts"
        ] = contem(
            renovate_json,
            "vulnerabilityalerts",
            "osv"
        )

        evidencias[
            "Lock File Maintenance"
        ] = contem(
            renovate_json,
            "lockfilemaintenance"
        )

else:

    evidencias["Renovate configurado"] = None

    evidencias[
        "Gerenciamento NuGet pelo Renovate"
    ] = None

    evidencias[
        "Gerenciamento Dockerfile pelo Renovate"
    ] = None

    evidencias[
        "Vulnerability Alerts"
    ] = None

    evidencias[
        "Lock File Maintenance"
    ] = None


# ---------------------------------------------------------
# 4. Renovate orquestrado no pipeline
# ---------------------------------------------------------

if pipeline:

    possui_imagem_renovate = contem(
        pipeline,
        "renovate/renovate"
    )

    possui_comando_renovate = contem(
        pipeline,
        "renovate $ci_project_path",
        "renovate ${ci_project_path}"
    )

    evidencias[
        "Job Renovate no pipeline"
    ] = (
        possui_imagem_renovate
        or possui_comando_renovate
    )


    evidencias[
        "Regra schedule para Renovate"
    ] = (
        evidencias[
            "Job Renovate no pipeline"
        ]
        and contem(
            pipeline,
            "schedule"
        )
    )


    evidencias[
        "Execução manual do Renovate"
    ] = (
        evidencias[
            "Job Renovate no pipeline"
        ]
        and contem(
            pipeline,
            "when: manual",
            "ci_pipeline_source == \"web\"",
            "ci_pipeline_source == 'web'"
        )
    )

else:

    evidencias[
        "Job Renovate no pipeline"
    ] = None

    evidencias[
        "Regra schedule para Renovate"
    ] = None

    evidencias[
        "Execução manual do Renovate"
    ] = None


# Configuração e job não comprovam execução efetiva
evidencias[
    "Execução efetiva do Renovate"
] = None


# ---------------------------------------------------------
# 5. Dockerfile
# ---------------------------------------------------------

if dockerfile:

    evidencias[
        "Dockerfile presente"
    ] = True

    # Detecta múltiplos FROM
    quantidade_from = len(
        re.findall(
            r"^\s*FROM\s+",
            dockerfile,
            flags=re.IGNORECASE | re.MULTILINE
        )
    )

    evidencias[
        "Docker multi-stage build"
    ] = (
        quantidade_from >= 2
    )

    evidencias[
        "Restore no Dockerfile"
    ] = contem(
        dockerfile,
        "dotnet restore"
    )

    evidencias[
        "Build no Dockerfile"
    ] = contem(
        dockerfile,
        "dotnet build"
    )

    evidencias[
        "Publish no Dockerfile"
    ] = contem(
        dockerfile,
        "dotnet publish"
    )

else:

    evidencias[
        "Dockerfile presente"
    ] = None

    evidencias[
        "Docker multi-stage build"
    ] = None

    evidencias[
        "Restore no Dockerfile"
    ] = None

    evidencias[
        "Build no Dockerfile"
    ] = None

    evidencias[
        "Publish no Dockerfile"
    ] = None


# ---------------------------------------------------------
# 6. Docker Compose
# ---------------------------------------------------------

if docker_compose:

    evidencias[
        "Docker Compose presente"
    ] = True

    evidencias[
        "Build definido no Docker Compose"
    ] = contem(
        docker_compose,
        "build:"
    )

    evidencias[
        "Política de restart"
    ] = contem(
        docker_compose,
        "restart:"
    )

    evidencias[
        "Rede externa no Docker Compose"
    ] = (
        contem(
            docker_compose,
            "external:"
        )
        and contem(
            docker_compose,
            "true"
        )
    )

    evidencias[
        "Reverse proxy configurado"
    ] = contem(
        docker_compose,
        "traefik",
        "nginx",
        "reverse proxy"
    )

else:

    evidencias[
        "Docker Compose presente"
    ] = None

    evidencias[
        "Build definido no Docker Compose"
    ] = None

    evidencias[
        "Política de restart"
    ] = None

    evidencias[
        "Rede externa no Docker Compose"
    ] = None

    evidencias[
        "Reverse proxy configurado"
    ] = None


# ---------------------------------------------------------
# 7. Program.cs / Observabilidade
# ---------------------------------------------------------

if program_cs:

    evidencias[
        "Sentry integrado em runtime"
    ] = contem(
        program_cs,
        "usesentry",
        "sentry"
    )

    evidencias[
        "Health Checks"
    ] = contem(
        program_cs,
        "addhealthchecks",
        "maphealthchecks"
    )

    evidencias[
        "Health Check de banco"
    ] = contem(
        program_cs,
        "addnpgsql",
        "npgsql",
        "healthcheck"
    )

    evidencias[
        "Health Check de API"
    ] = (
        contem(
            program_cs,
            "addurlgroup",
            "httpclient",
            "healthcheck"
        )
        and contem(
            program_cs,
            "health"
        )
    )

    evidencias[
        "Console Logging"
    ] = contem(
        program_cs,
        "addconsole",
        "console"
    )

else:

    evidencias[
        "Sentry integrado em runtime"
    ] = None

    evidencias[
        "Health Checks"
    ] = None

    evidencias[
        "Health Check de banco"
    ] = None

    evidencias[
        "Health Check de API"
    ] = None

    evidencias[
        "Console Logging"
    ] = None


# ---------------------------------------------------------
# 8. appsettings
# ---------------------------------------------------------

if appsettings:

    evidencias[
        "appsettings presente"
    ] = True

    evidencias[
        "Sentry configurado"
    ] = contem(
        appsettings,
        "sentry",
        "dsn"
    )

    evidencias[
        "Matomo configurado"
    ] = contem(
        appsettings,
        "matomo"
    )

    evidencias[
        "PostgreSQL configurado"
    ] = contem(
        appsettings,
        "postgres",
        "npgsql"
    )

    evidencias[
        "Configuração externalizada"
    ] = contem(
        appsettings,
        "environment",
        "connectionstrings",
        "configuration"
    )

else:

    evidencias[
        "appsettings presente"
    ] = None

    evidencias[
        "Sentry configurado"
    ] = None

    evidencias[
        "Matomo configurado"
    ] = None

    evidencias[
        "PostgreSQL configurado"
    ] = None

    evidencias[
        "Configuração externalizada"
    ] = None


# ---------------------------------------------------------
# 9. .editorconfig
# ---------------------------------------------------------

if editorconfig:

    evidencias[
        ".editorconfig presente"
    ] = True

    evidencias[
        "Regras de analyzer declaradas"
    ] = contem(
        editorconfig,
        "dotnet_diagnostic",
        "severity"
    )

    evidencias[
        "Regras de segurança declaradas"
    ] = contem(
        editorconfig,
        "security",
        "ca2",
        "s"
    )

else:

    evidencias[
        ".editorconfig presente"
    ] = None

    evidencias[
        "Regras de analyzer declaradas"
    ] = None

    evidencias[
        "Regras de segurança declaradas"
    ] = None


# ---------------------------------------------------------
# 10. .env.example
# ---------------------------------------------------------

if env_example:

    evidencias[
        ".env.example presente"
    ] = True

    evidencias[
        "Parametrização de ambientes"
    ] = contem(
        env_example,
        "environment",
        "env",
        "stage",
        "server",
        "host"
    )

else:

    evidencias[
        ".env.example presente"
    ] = None

    evidencias[
        "Parametrização de ambientes"
    ] = None


# ---------------------------------------------------------
# 11. .env.io.example
# ---------------------------------------------------------

if env_io_example:

    evidencias[
        ".env.io.example presente"
    ] = True

    # Se já havia evidência anterior, preserva True
    evidencia_parametrizacao = contem(
        env_io_example,
        "environment",
        "stage",
        "server",
        "version",
        "deployer",
        "project"
    )

    valor_atual = evidencias.get(
        "Parametrização de ambientes"
    )

    if valor_atual is True:
        evidencias[
            "Parametrização de ambientes"
        ] = True

    elif valor_atual is False:
        evidencias[
            "Parametrização de ambientes"
        ] = evidencia_parametrizacao

    else:
        evidencias[
            "Parametrização de ambientes"
        ] = evidencia_parametrizacao

else:

    evidencias[
        ".env.io.example presente"
    ] = None


# ---------------------------------------------------------
# 12. Scripts
# ---------------------------------------------------------

if scripts:

    evidencias[
        "Scripts presentes"
    ] = True

    conteudo_scripts = "\n".join(
        str(conteudo)
        for conteudo in scripts.values()
    )

    nomes_scripts = "\n".join(
        str(nome)
        for nome in scripts.keys()
    )

    conteudo_total_scripts = (
        nomes_scripts
        + "\n"
        + conteudo_scripts
    )

    evidencias[
        "Script de inicialização de banco"
    ] = contem(
        conteudo_total_scripts,
        "init.sql",
        "create table",
        "create schema"
    )

    evidencias[
        "Script de seed"
    ] = contem(
        conteudo_total_scripts,
        "seed.sql",
        "insert into"
    )

    evidencias[
        "Dump de banco presente"
    ] = contem(
        conteudo_total_scripts,
        "pg_dump",
        ".dump",
        "postgresql database dump"
    )

    evidencias[
        "Automação de ambiente local"
    ] = contem(
        conteudo_total_scripts,
        "docker compose",
        "docker-compose",
        "start-db",
        "init.sql",
        "localhost"
    )

else:

    evidencias[
        "Scripts presentes"
    ] = None

    evidencias[
        "Script de inicialização de banco"
    ] = None

    evidencias[
        "Script de seed"
    ] = None

    evidencias[
        "Dump de banco presente"
    ] = None

    evidencias[
        "Automação de ambiente local"
    ] = None


# ---------------------------------------------------------
# 13. Resumo tecnológico
# ---------------------------------------------------------

tecnologia = {
    "target_framework": target_framework,
    "sdk_pipeline": sdk_pipeline
}


# ---------------------------------------------------------
# 14. Resultado
# ---------------------------------------------------------

print("EVIDÊNCIAS EXTRAÍDAS\n")

print(
    "Target Framework:",
    target_framework
)

print(
    "SDK Pipeline:",
    sdk_pipeline
)

print()


# ---------------------------------------------------------
# 15. Exibição tri-state
# ---------------------------------------------------------

for nome, valor in evidencias.items():

    if valor is True:
        simbolo = "✓"

    elif valor is False:
        simbolo = "—"

    else:
        simbolo = "?"

    print(
        f"{simbolo} {nome}"
    )
```

```python
# ---------------------------------------------------------
# ARTIFACT IDENTIFICATION AND VALIDATION
# ---------------------------------------------------------

if not arquivos:
    raise ValueError(
        "Nenhum arquivo foi carregado. "
        "Execute primeiro a etapa Artifact Upload."
    )


# ---------------------------------------------------------
# Reinicia variáveis de identificação
# ---------------------------------------------------------

pipeline = None
pipeline_nome = None

csproj = None
csproj_nome = None

renovate_json = None
renovate_json_nome = None

renovate_setup = None
renovate_setup_nome = None

dockerfile = None
dockerfile_nome = None

docker_compose = None
docker_compose_nome = None

program_cs = None
program_cs_nome = None

appsettings = None
appsettings_nome = None

editorconfig = None
editorconfig_nome = None

env_example = None
env_example_nome = None

env_io_example = None
env_io_example_nome = None

scripts = {}
scripts_nomes = []

validacao_artefatos = []


# ---------------------------------------------------------
# Estados formais de validação
#
# None  = artefato não fornecido / não validável
# True  = artefato fornecido e validado
# False = artefato fornecido, porém inválido
# ---------------------------------------------------------

pipeline_valido = None
csproj_valido = None
renovate_json_valido = None
renovate_setup_valido = None
dockerfile_valido = None
docker_compose_valido = None
program_cs_valido = None
appsettings_valido = None
editorconfig_valido = None
env_example_valido = None
env_io_example_valido = None

scripts_validos = {}


# ---------------------------------------------------------
# Normalização de nomes
# ---------------------------------------------------------

def nome_base(nome):
    return (
        nome
        .replace("\\", "/")
        .split("/")[-1]
        .strip()
    )


def normalizar_nome(nome):
    nome = nome_base(nome)

    # Remove sufixos adicionados pelo Colab:
    # arquivo (1).json -> arquivo.json
    # Dockerfile (2)   -> Dockerfile
    nome = re.sub(
        r"\s*\(\d+\)(?=\.|$)",
        "",
        nome
    )

    return nome.lower().strip()


# ---------------------------------------------------------
# Seleção de artefato único
# ---------------------------------------------------------

def selecionar_unico(
    candidatos,
    descricao
):
    if not candidatos:
        return None, None

    if len(candidatos) > 1:

        nomes = ", ".join(
            sorted(
                nome
                for nome, _ in candidatos
            )
        )

        raise ValueError(
            f"Mais de um arquivo foi identificado como "
            f"{descricao}: {nomes}. "
            "Forneça apenas um artefato desse tipo por versão."
        )

    return (
        candidatos[0][1],
        candidatos[0][0]
    )


# ---------------------------------------------------------
# Conversão segura para texto
# ---------------------------------------------------------

def como_texto(conteudo):
    if conteudo is None:
        return None

    if isinstance(
        conteudo,
        bytes
    ):
        return conteudo.decode(
            "utf-8-sig",
            errors="replace"
        )

    texto = str(
        conteudo
    )

    # Remove BOM caso o conteúdo já seja string.
    return texto.lstrip(
        "\ufeff"
    )


# ---------------------------------------------------------
# Candidatos
# ---------------------------------------------------------

candidatos_pipeline = []
candidatos_csproj = []
candidatos_renovate = []
candidatos_renovate_setup = []
candidatos_dockerfile = []
candidatos_docker_compose = []
candidatos_program = []
candidatos_appsettings = []
candidatos_editorconfig = []
candidatos_env_example = []
candidatos_env_io_example = []

extensoes_script = {
    ".sql",
    ".ps1",
    ".sh",
    ".bat",
    ".cmd"
}


# ---------------------------------------------------------
# Identificação dos arquivos
# ---------------------------------------------------------

for nome, conteudo in arquivos.items():

    base = nome_base(
        nome
    )

    normalizado = normalizar_nome(
        nome
    )


    # GitLab CI
    if normalizado in {
        ".gitlab-ci.yml",
        ".gitlab-ci.yaml",
        "gitlab-ci.yml",
        "gitlab-ci.yaml"
    }:

        candidatos_pipeline.append(
            (
                base,
                conteudo
            )
        )

        continue


    # Projeto .NET
    if normalizado.endswith(
        ".csproj"
    ):

        candidatos_csproj.append(
            (
                base,
                conteudo
            )
        )

        continue


    # Renovate
    if normalizado == "renovate.json":

        candidatos_renovate.append(
            (
                base,
                conteudo
            )
        )

        continue


    if normalizado == "renovate_setup.md":

        candidatos_renovate_setup.append(
            (
                base,
                conteudo
            )
        )

        continue


    # Dockerfile
    if normalizado == "dockerfile":

        candidatos_dockerfile.append(
            (
                base,
                conteudo
            )
        )

        continue


    # Docker Compose
    if normalizado in {
        "docker-compose.yml",
        "docker-compose.yaml",
        "compose.yml",
        "compose.yaml"
    }:

        candidatos_docker_compose.append(
            (
                base,
                conteudo
            )
        )

        continue


    # Program.cs
    if normalizado == "program.cs":

        candidatos_program.append(
            (
                base,
                conteudo
            )
        )

        continue


    # appsettings.json
    if normalizado == "appsettings.json":

        candidatos_appsettings.append(
            (
                base,
                conteudo
            )
        )

        continue


    # .editorconfig
    if normalizado == ".editorconfig":

        candidatos_editorconfig.append(
            (
                base,
                conteudo
            )
        )

        continue


    # .env.io.example
    if normalizado in {
        ".env.io.example",
        "env.io.example"
    }:

        candidatos_env_io_example.append(
            (
                base,
                conteudo
            )
        )

        continue


    # .env.example
    if normalizado in {
        ".env.example",
        "env.example"
    }:

        candidatos_env_example.append(
            (
                base,
                conteudo
            )
        )

        continue


    # Scripts
    extensao = Path(
        normalizado
    ).suffix

    if extensao in extensoes_script:

        scripts[
            base
        ] = como_texto(
            conteudo
        )

        scripts_nomes.append(
            base
        )


# ---------------------------------------------------------
# Seleção dos artefatos
# ---------------------------------------------------------

pipeline, pipeline_nome = selecionar_unico(
    candidatos_pipeline,
    "pipeline GitLab CI"
)

csproj, csproj_nome = selecionar_unico(
    candidatos_csproj,
    "arquivo .csproj"
)

renovate_json, renovate_json_nome = selecionar_unico(
    candidatos_renovate,
    "renovate.json"
)

renovate_setup, renovate_setup_nome = selecionar_unico(
    candidatos_renovate_setup,
    "RENOVATE_SETUP.md"
)

dockerfile, dockerfile_nome = selecionar_unico(
    candidatos_dockerfile,
    "Dockerfile"
)

docker_compose, docker_compose_nome = selecionar_unico(
    candidatos_docker_compose,
    "Docker Compose"
)

program_cs, program_cs_nome = selecionar_unico(
    candidatos_program,
    "Program.cs"
)

appsettings, appsettings_nome = selecionar_unico(
    candidatos_appsettings,
    "appsettings.json"
)

editorconfig, editorconfig_nome = selecionar_unico(
    candidatos_editorconfig,
    ".editorconfig"
)

env_example, env_example_nome = selecionar_unico(
    candidatos_env_example,
    ".env.example"
)

env_io_example, env_io_example_nome = selecionar_unico(
    candidatos_env_io_example,
    ".env.io.example"
)


# ---------------------------------------------------------
# Conversão para texto
# ---------------------------------------------------------

pipeline = como_texto(
    pipeline
)

csproj = como_texto(
    csproj
)

renovate_json = como_texto(
    renovate_json
)

renovate_setup = como_texto(
    renovate_setup
)

dockerfile = como_texto(
    dockerfile
)

docker_compose = como_texto(
    docker_compose
)

program_cs = como_texto(
    program_cs
)

appsettings = como_texto(
    appsettings
)

editorconfig = como_texto(
    editorconfig
)

env_example = como_texto(
    env_example
)

env_io_example = como_texto(
    env_io_example
)


# ---------------------------------------------------------
# Funções auxiliares de validação
# ---------------------------------------------------------

def arquivo_nao_vazio(
    conteudo
):
    return (
        conteudo is not None
        and bool(
            conteudo.strip()
        )
    )


def validar_env(
    conteudo,
    descricao
):
    if conteudo is None:
        return None

    if not arquivo_nao_vazio(
        conteudo
    ):
        validacao_artefatos.append(
            f"✗ {descricao}: arquivo vazio."
        )

        return False

    linhas_invalidas = []

    for numero, linha in enumerate(
        conteudo.splitlines(),
        start=1
    ):

        linha = linha.strip()

        if (
            not linha
            or linha.startswith("#")
        ):
            continue

        if "=" not in linha:
            linhas_invalidas.append(
                numero
            )

    if linhas_invalidas:

        validacao_artefatos.append(
            f"⚠ {descricao}: "
            f"{len(linhas_invalidas)} linha(s) "
            "sem formato CHAVE=VALOR."
        )

        # O arquivo continua legível e utilizável
        # para extração de evidências.
        return True

    validacao_artefatos.append(
        f"✓ {descricao} estruturalmente válido."
    )

    return True


# ---------------------------------------------------------
# 1. Pipeline YAML
# ---------------------------------------------------------

if pipeline is not None:

    if not arquivo_nao_vazio(
        pipeline
    ):

        pipeline_valido = False

        validacao_artefatos.append(
            "✗ Pipeline: arquivo vazio."
        )

    else:

        try:

            pipeline_yaml = yaml.safe_load(
                pipeline
            )

            if not isinstance(
                pipeline_yaml,
                dict
            ):
                raise ValueError(
                    "estrutura YAML raiz não é um mapeamento"
                )

            pipeline_valido = True

            validacao_artefatos.append(
                "✓ Pipeline YAML válido."
            )

        except Exception as erro:

            pipeline_valido = False

            validacao_artefatos.append(
                f"✗ Pipeline YAML inválido: {erro}"
            )


# ---------------------------------------------------------
# 2. .csproj XML
# ---------------------------------------------------------

if csproj is not None:

    if not arquivo_nao_vazio(
        csproj
    ):

        csproj_valido = False

        validacao_artefatos.append(
            "✗ .csproj: arquivo vazio."
        )

    else:

        try:

            ET.fromstring(
                csproj
            )

            csproj_valido = True

            validacao_artefatos.append(
                "✓ .csproj XML válido."
            )

        except ET.ParseError as erro:

            csproj_valido = False

            validacao_artefatos.append(
                f"✗ .csproj XML inválido: {erro}"
            )


# ---------------------------------------------------------
# 3. renovate.json
# ---------------------------------------------------------

if renovate_json is not None:

    if not arquivo_nao_vazio(
        renovate_json
    ):

        renovate_json_valido = False

        validacao_artefatos.append(
            "✗ renovate.json: arquivo vazio."
        )

    else:

        try:

            renovate_data_validacao = json.loads(
                renovate_json
            )

            if not isinstance(
                renovate_data_validacao,
                dict
            ):
                raise ValueError(
                    "raiz JSON não é um objeto"
                )

            renovate_json_valido = True

            validacao_artefatos.append(
                "✓ renovate.json válido."
            )

        except Exception as erro:

            renovate_json_valido = False

            validacao_artefatos.append(
                f"✗ renovate.json inválido: {erro}"
            )


# ---------------------------------------------------------
# 4. RENOVATE_SETUP.md
# ---------------------------------------------------------

if renovate_setup is not None:

    if arquivo_nao_vazio(
        renovate_setup
    ):

        renovate_setup_valido = True

        validacao_artefatos.append(
            "✓ RENOVATE_SETUP.md legível e não vazio."
        )

    else:

        renovate_setup_valido = False

        validacao_artefatos.append(
            "✗ RENOVATE_SETUP.md: arquivo vazio."
        )


# ---------------------------------------------------------
# 5. Dockerfile
# ---------------------------------------------------------

if dockerfile is not None:

    if not arquivo_nao_vazio(
        dockerfile
    ):

        dockerfile_valido = False

        validacao_artefatos.append(
            "✗ Dockerfile: arquivo vazio."
        )

    else:

        possui_from = bool(
            re.search(
                r"^\s*FROM\s+\S+",
                dockerfile,
                flags=(
                    re.IGNORECASE
                    | re.MULTILINE
                )
            )
        )

        if possui_from:

            dockerfile_valido = True

            validacao_artefatos.append(
                "✓ Dockerfile estruturalmente válido "
                "(instrução FROM identificada)."
            )

        else:

            dockerfile_valido = False

            validacao_artefatos.append(
                "✗ Dockerfile sem instrução FROM identificada."
            )


# ---------------------------------------------------------
# 6. Docker Compose
# ---------------------------------------------------------

if docker_compose is not None:

    if not arquivo_nao_vazio(
        docker_compose
    ):

        docker_compose_valido = False

        validacao_artefatos.append(
            "✗ Docker Compose: arquivo vazio."
        )

    else:

        try:

            docker_compose_obj = yaml.safe_load(
                docker_compose
            )

            if not isinstance(
                docker_compose_obj,
                dict
            ):
                raise ValueError(
                    "estrutura YAML raiz não é um mapeamento"
                )

            if "services" not in docker_compose_obj:

                docker_compose_valido = False

                validacao_artefatos.append(
                    "✗ Docker Compose YAML válido, "
                    "mas sem seção 'services'."
                )

            else:

                docker_compose_valido = True

                validacao_artefatos.append(
                    "✓ Docker Compose YAML válido "
                    "e seção 'services' identificada."
                )

        except Exception as erro:

            docker_compose_valido = False

            validacao_artefatos.append(
                f"✗ Docker Compose YAML inválido: {erro}"
            )


# ---------------------------------------------------------
# 7. Program.cs
# ---------------------------------------------------------

if program_cs is not None:

    if not arquivo_nao_vazio(
        program_cs
    ):

        program_cs_valido = False

        validacao_artefatos.append(
            "✗ Program.cs: arquivo vazio."
        )

    else:

        indicadores_cs = (
            "WebApplication",
            "CreateBuilder",
            "builder.",
            "Build("
        )

        program_cs_valido = any(
            indicador.lower()
            in program_cs.lower()
            for indicador in indicadores_cs
        )

        if program_cs_valido:

            validacao_artefatos.append(
                "✓ Program.cs legível e com estrutura "
                "compatível com inicialização de aplicação .NET."
            )

        else:

            validacao_artefatos.append(
                "⚠ Program.cs legível, mas a estrutura "
                "esperada de inicialização .NET "
                "não foi identificada."
            )


# ---------------------------------------------------------
# 8. appsettings.json
# ---------------------------------------------------------

if appsettings is not None:

    if not arquivo_nao_vazio(
        appsettings
    ):

        appsettings_valido = False

        validacao_artefatos.append(
            "✗ appsettings.json: arquivo vazio."
        )

    else:

        try:

            appsettings_obj = json.loads(
                appsettings
            )

            if not isinstance(
                appsettings_obj,
                dict
            ):
                raise ValueError(
                    "raiz JSON não é um objeto"
                )

            appsettings_valido = True

            validacao_artefatos.append(
                "✓ appsettings.json válido."
            )

        except Exception as erro:

            appsettings_valido = False

            validacao_artefatos.append(
                f"✗ appsettings.json inválido: {erro}"
            )


# ---------------------------------------------------------
# 9. .editorconfig
# ---------------------------------------------------------

if editorconfig is not None:

    if not arquivo_nao_vazio(
        editorconfig
    ):

        editorconfig_valido = False

        validacao_artefatos.append(
            "✗ .editorconfig: arquivo vazio."
        )

    else:

        possui_secao = bool(
            re.search(
                r"^\s*\[[^\]]+\]\s*$",
                editorconfig,
                flags=re.MULTILINE
            )
        )

        editorconfig_valido = possui_secao

        if possui_secao:

            validacao_artefatos.append(
                "✓ .editorconfig estruturalmente válido "
                "(seção de configuração identificada)."
            )

        else:

            validacao_artefatos.append(
                "⚠ .editorconfig legível, "
                "mas nenhuma seção foi identificada."
            )


# ---------------------------------------------------------
# 10. Arquivos de ambiente
# ---------------------------------------------------------

env_example_valido = validar_env(
    env_example,
    ".env.example"
)

env_io_example_valido = validar_env(
    env_io_example,
    ".env.io.example"
)


# ---------------------------------------------------------
# 11. Scripts
# ---------------------------------------------------------

for script_nome in sorted(
    scripts_nomes
):

    conteudo_script = scripts.get(
        script_nome
    )

    if arquivo_nao_vazio(
        conteudo_script
    ):

        scripts_validos[
            script_nome
        ] = True

        extensao = Path(
            normalizar_nome(
                script_nome
            )
        ).suffix

        tipo_script = {
            ".sql": "SQL",
            ".ps1": "PowerShell",
            ".sh": "Shell",
            ".bat": "Batch",
            ".cmd": "Command"
        }.get(
            extensao,
            "script"
        )

        validacao_artefatos.append(
            f"✓ {script_nome}: "
            f"script {tipo_script} legível e não vazio."
        )

    else:

        scripts_validos[
            script_nome
        ] = False

        validacao_artefatos.append(
            f"✗ {script_nome}: arquivo vazio."
        )


# ---------------------------------------------------------
# Resultado da identificação
# ---------------------------------------------------------

print(
    "ARTEFATOS IDENTIFICADOS\n"
)


def imprimir_artefato(
    descricao,
    nome
):
    if nome:

        print(
            f"✓ {descricao}: {nome}"
        )

    else:

        print(
            f"? {descricao}: não fornecido"
        )


imprimir_artefato(
    "Pipeline",
    pipeline_nome
)

imprimir_artefato(
    ".csproj",
    csproj_nome
)

imprimir_artefato(
    "renovate.json",
    renovate_json_nome
)

imprimir_artefato(
    "RENOVATE_SETUP.md",
    renovate_setup_nome
)

imprimir_artefato(
    "Dockerfile",
    dockerfile_nome
)

imprimir_artefato(
    "Docker Compose",
    docker_compose_nome
)

imprimir_artefato(
    "Program.cs",
    program_cs_nome
)

imprimir_artefato(
    "appsettings.json",
    appsettings_nome
)

imprimir_artefato(
    ".editorconfig",
    editorconfig_nome
)

imprimir_artefato(
    ".env.example",
    env_example_nome
)

imprimir_artefato(
    ".env.io.example",
    env_io_example_nome
)


if scripts_nomes:

    print(
        f"✓ Scripts: "
        f"{len(scripts_nomes)} arquivo(s)"
    )

    for nome in sorted(
        scripts_nomes
    ):
        print(
            f"    - {nome}"
        )

else:

    print(
        "? Scripts: não fornecidos"
    )


# ---------------------------------------------------------
# Resultado da validação
# ---------------------------------------------------------

print(
    "\nVALIDAÇÃO\n"
)

if validacao_artefatos:

    for mensagem in validacao_artefatos:

        print(
            mensagem
        )

else:

    print(
        "Nenhum artefato sujeito a validação "
        "foi fornecido."
    )


# ---------------------------------------------------------
# Estado técnico da validação
# ---------------------------------------------------------

print(
    "\nESTADO DA VALIDAÇÃO\n"
)

estados_validacao = {
    "Pipeline": pipeline_valido,
    ".csproj": csproj_valido,
    "renovate.json": renovate_json_valido,
    "RENOVATE_SETUP.md": renovate_setup_valido,
    "Dockerfile": dockerfile_valido,
    "Docker Compose": docker_compose_valido,
    "Program.cs": program_cs_valido,
    "appsettings.json": appsettings_valido,
    ".editorconfig": editorconfig_valido,
    ".env.example": env_example_valido,
    ".env.io.example": env_io_example_valido
}


for artefato, estado in estados_validacao.items():

    if estado is True:
        simbolo = "✓"

    elif estado is False:
        simbolo = "✗"

    else:
        simbolo = "?"

    print(
        f"{simbolo} {artefato}"
    )
```

```python
# ---------------------------------------------------------
# PIPELINE MATURITY CLASSIFICATION
# ---------------------------------------------------------

def classificar_maturidade(e):

    if not e:

        return {
            "nivel": None,
            "classificacao": "Indeterminado",
            "motivo": (
                "Nenhuma evidência foi extraída."
            )
        }


    gitlab_ci = e.get(
        "GitLab CI como código"
    )

    restore = e.get(
        "`dotnet restore` no CI"
    )

    build = e.get(
        "Build Release no CI"
    )

    sonar = e.get(
        "SonarQube"
    )

    quality_gate_wait = e.get(
        "Espera pelo Quality Gate"
    )

    quality_gate_blocking = e.get(
        "Quality Gate bloqueante"
    )

    testes = e.get(
        "Testes automatizados"
    )

    deploy = e.get(
        "Deploy automatizado"
    )

    promocao = e.get(
        "Promoção automática"
    )

    rollback = e.get(
        "Rollback automatizado"
    )


    # -----------------------------------------------------
    # Evidência insuficiente
    # -----------------------------------------------------

    if all(
        valor is None
        for valor in (
            gitlab_ci,
            restore,
            build,
            sonar
        )
    ):

        return {
            "nivel": None,
            "classificacao": "Indeterminado",
            "motivo": (
                "O pipeline não está disponível ou não pôde "
                "ser validado."
            )
        }


    # -----------------------------------------------------
    # CI estruturada
    # -----------------------------------------------------

    ci_estruturada = (
        gitlab_ci is True
        and restore is True
        and build is True
    )

    qualidade = (
        sonar is True
        or quality_gate_wait is True
    )


    # -----------------------------------------------------
    # Nível 4
    # -----------------------------------------------------

    if (
        ci_estruturada
        and qualidade
        and testes is True
        and quality_gate_blocking is True
        and deploy is True
        and promocao is True
        and rollback is True
    ):

        return {
            "nivel": 4,
            "classificacao": "Avançado/Consolidado",
            "motivo": (
                "O pipeline apresenta CI estruturada, testes "
                "automatizados, controle de qualidade bloqueante, "
                "deploy, promoção e rollback automatizados."
            )
        }


    # -----------------------------------------------------
    # Nível 3
    # -----------------------------------------------------

    if (
        ci_estruturada
        and qualidade
    ):

        return {
            "nivel": 3,
            "classificacao": "Parcial/Emergente",
            "motivo": (
                "O pipeline apresenta integração contínua "
                "estruturada e controles de qualidade, mas "
                "a automação de ponta a ponta permanece incompleta."
            )
        }


    # -----------------------------------------------------
    # Nível 2
    # -----------------------------------------------------

    if (
        gitlab_ci is True
        and (
            restore is True
            or build is True
            or sonar is True
        )
    ):

        return {
            "nivel": 2,
            "classificacao": "Inicial/Ad hoc",
            "motivo": (
                "Existe pipeline versionado com automações "
                "pontuais, mas sem CI estruturada completa."
            )
        }


    # -----------------------------------------------------
    # Nível 1
    # -----------------------------------------------------

    if gitlab_ci is True:

        return {
            "nivel": 1,
            "classificacao": "Ausente/Tradicional",
            "motivo": (
                "O pipeline existe, mas não foram identificadas "
                "automações relevantes segundo os critérios definidos."
            )
        }


    return {
        "nivel": None,
        "classificacao": "Indeterminado",
        "motivo": (
            "As evidências disponíveis são insuficientes."
        )
    }


# ---------------------------------------------------------
# Executa a classificação
# ---------------------------------------------------------

resultado_maturidade = classificar_maturidade(
    evidencias
)


# ---------------------------------------------------------
# Resultado
# ---------------------------------------------------------

print("CLASSIFICAÇÃO DE MATURIDADE\n")

print(
    "Nível:",
    resultado_maturidade["nivel"]
)

print(
    "Classificação:",
    resultado_maturidade["classificacao"]
)

print(
    "Motivo:",
    resultado_maturidade["motivo"]
)
```

```python
# ---------------------------------------------------------
# ARTIFACT IDENTIFICATION AND VALIDATION
# ---------------------------------------------------------

if not arquivos:
    raise ValueError(
        "Nenhum arquivo foi carregado. "
        "Execute primeiro a etapa Artifact Upload."
    )


# ---------------------------------------------------------
# Reinicia variáveis de identificação
# ---------------------------------------------------------

pipeline = None
pipeline_nome = None

csproj = None
csproj_nome = None

renovate_json = None
renovate_json_nome = None

renovate_setup = None
renovate_setup_nome = None

dockerfile = None
dockerfile_nome = None

docker_compose = None
docker_compose_nome = None

program_cs = None
program_cs_nome = None

appsettings = None
appsettings_nome = None

editorconfig = None
editorconfig_nome = None

env_example = None
env_example_nome = None

env_io_example = None
env_io_example_nome = None

scripts = {}
scripts_nomes = []

validacao_artefatos = []


# ---------------------------------------------------------
# Estados formais de validação
#
# None  = artefato não fornecido / não validável
# True  = artefato fornecido e validado
# False = artefato fornecido, porém inválido
# ---------------------------------------------------------

pipeline_valido = None
csproj_valido = None
renovate_json_valido = None
renovate_setup_valido = None
dockerfile_valido = None
docker_compose_valido = None
program_cs_valido = None
appsettings_valido = None
editorconfig_valido = None
env_example_valido = None
env_io_example_valido = None

scripts_validos = {}


# ---------------------------------------------------------
# Normalização de nomes
# ---------------------------------------------------------

def nome_base(nome):
    return (
        nome
        .replace("\\", "/")
        .split("/")[-1]
        .strip()
    )


def normalizar_nome(nome):
    nome = nome_base(nome)

    # Remove sufixos adicionados pelo Colab:
    # arquivo (1).json -> arquivo.json
    # Dockerfile (2)   -> Dockerfile
    nome = re.sub(
        r"\s*\(\d+\)(?=\.|$)",
        "",
        nome
    )

    return nome.lower().strip()


# ---------------------------------------------------------
# Seleção de artefato único
# ---------------------------------------------------------

def selecionar_unico(
    candidatos,
    descricao
):
    if not candidatos:
        return None, None

    if len(candidatos) > 1:

        nomes = ", ".join(
            sorted(
                nome
                for nome, _ in candidatos
            )
        )

        raise ValueError(
            f"Mais de um arquivo foi identificado como "
            f"{descricao}: {nomes}. "
            "Forneça apenas um artefato desse tipo por versão."
        )

    return (
        candidatos[0][1],
        candidatos[0][0]
    )


# ---------------------------------------------------------
# Conversão segura para texto
# ---------------------------------------------------------

def como_texto(conteudo):
    if conteudo is None:
        return None

    if isinstance(
        conteudo,
        bytes
    ):
        return conteudo.decode(
            "utf-8-sig",
            errors="replace"
        )

    texto = str(
        conteudo
    )

    # Remove BOM caso o conteúdo já seja string.
    return texto.lstrip(
        "\ufeff"
    )


# ---------------------------------------------------------
# Candidatos
# ---------------------------------------------------------

candidatos_pipeline = []
candidatos_csproj = []
candidatos_renovate = []
candidatos_renovate_setup = []
candidatos_dockerfile = []
candidatos_docker_compose = []
candidatos_program = []
candidatos_appsettings = []
candidatos_editorconfig = []
candidatos_env_example = []
candidatos_env_io_example = []

extensoes_script = {
    ".sql",
    ".ps1",
    ".sh",
    ".bat",
    ".cmd"
}


# ---------------------------------------------------------
# Identificação dos arquivos
# ---------------------------------------------------------

for nome, conteudo in arquivos.items():

    base = nome_base(
        nome
    )

    normalizado = normalizar_nome(
        nome
    )


    # GitLab CI
    if normalizado in {
        ".gitlab-ci.yml",
        ".gitlab-ci.yaml",
        "gitlab-ci.yml",
        "gitlab-ci.yaml"
    }:

        candidatos_pipeline.append(
            (
                base,
                conteudo
            )
        )

        continue


    # Projeto .NET
    if normalizado.endswith(
        ".csproj"
    ):

        candidatos_csproj.append(
            (
                base,
                conteudo
            )
        )

        continue


    # Renovate
    if normalizado == "renovate.json":

        candidatos_renovate.append(
            (
                base,
                conteudo
            )
        )

        continue


    if normalizado == "renovate_setup.md":

        candidatos_renovate_setup.append(
            (
                base,
                conteudo
            )
        )

        continue


    # Dockerfile
    if normalizado == "dockerfile":

        candidatos_dockerfile.append(
            (
                base,
                conteudo
            )
        )

        continue


    # Docker Compose
    if normalizado in {
        "docker-compose.yml",
        "docker-compose.yaml",
        "compose.yml",
        "compose.yaml"
    }:

        candidatos_docker_compose.append(
            (
                base,
                conteudo
            )
        )

        continue


    # Program.cs
    if normalizado == "program.cs":

        candidatos_program.append(
            (
                base,
                conteudo
            )
        )

        continue


    # appsettings.json
    if normalizado == "appsettings.json":

        candidatos_appsettings.append(
            (
                base,
                conteudo
            )
        )

        continue


    # .editorconfig
    if normalizado == ".editorconfig":

        candidatos_editorconfig.append(
            (
                base,
                conteudo
            )
        )

        continue


    # .env.io.example
    if normalizado in {
        ".env.io.example",
        "env.io.example"
    }:

        candidatos_env_io_example.append(
            (
                base,
                conteudo
            )
        )

        continue


    # .env.example
    if normalizado in {
        ".env.example",
        "env.example"
    }:

        candidatos_env_example.append(
            (
                base,
                conteudo
            )
        )

        continue


    # Scripts
    extensao = Path(
        normalizado
    ).suffix

    if extensao in extensoes_script:

        scripts[
            base
        ] = como_texto(
            conteudo
        )

        scripts_nomes.append(
            base
        )


# ---------------------------------------------------------
# Seleção dos artefatos
# ---------------------------------------------------------

pipeline, pipeline_nome = selecionar_unico(
    candidatos_pipeline,
    "pipeline GitLab CI"
)

csproj, csproj_nome = selecionar_unico(
    candidatos_csproj,
    "arquivo .csproj"
)

renovate_json, renovate_json_nome = selecionar_unico(
    candidatos_renovate,
    "renovate.json"
)

renovate_setup, renovate_setup_nome = selecionar_unico(
    candidatos_renovate_setup,
    "RENOVATE_SETUP.md"
)

dockerfile, dockerfile_nome = selecionar_unico(
    candidatos_dockerfile,
    "Dockerfile"
)

docker_compose, docker_compose_nome = selecionar_unico(
    candidatos_docker_compose,
    "Docker Compose"
)

program_cs, program_cs_nome = selecionar_unico(
    candidatos_program,
    "Program.cs"
)

appsettings, appsettings_nome = selecionar_unico(
    candidatos_appsettings,
    "appsettings.json"
)

editorconfig, editorconfig_nome = selecionar_unico(
    candidatos_editorconfig,
    ".editorconfig"
)

env_example, env_example_nome = selecionar_unico(
    candidatos_env_example,
    ".env.example"
)

env_io_example, env_io_example_nome = selecionar_unico(
    candidatos_env_io_example,
    ".env.io.example"
)


# ---------------------------------------------------------
# Conversão para texto
# ---------------------------------------------------------

pipeline = como_texto(
    pipeline
)

csproj = como_texto(
    csproj
)

renovate_json = como_texto(
    renovate_json
)

renovate_setup = como_texto(
    renovate_setup
)

dockerfile = como_texto(
    dockerfile
)

docker_compose = como_texto(
    docker_compose
)

program_cs = como_texto(
    program_cs
)

appsettings = como_texto(
    appsettings
)

editorconfig = como_texto(
    editorconfig
)

env_example = como_texto(
    env_example
)

env_io_example = como_texto(
    env_io_example
)


# ---------------------------------------------------------
# Funções auxiliares de validação
# ---------------------------------------------------------

def arquivo_nao_vazio(
    conteudo
):
    return (
        conteudo is not None
        and bool(
            conteudo.strip()
        )
    )


def validar_env(
    conteudo,
    descricao
):
    if conteudo is None:
        return None

    if not arquivo_nao_vazio(
        conteudo
    ):
        validacao_artefatos.append(
            f"✗ {descricao}: arquivo vazio."
        )

        return False

    linhas_invalidas = []

    for numero, linha in enumerate(
        conteudo.splitlines(),
        start=1
    ):

        linha = linha.strip()

        if (
            not linha
            or linha.startswith("#")
        ):
            continue

        if "=" not in linha:
            linhas_invalidas.append(
                numero
            )

    if linhas_invalidas:

        validacao_artefatos.append(
            f"⚠ {descricao}: "
            f"{len(linhas_invalidas)} linha(s) "
            "sem formato CHAVE=VALOR."
        )

        # O arquivo continua legível e utilizável
        # para extração de evidências.
        return True

    validacao_artefatos.append(
        f"✓ {descricao} estruturalmente válido."
    )

    return True


# ---------------------------------------------------------
# 1. Pipeline YAML
# ---------------------------------------------------------

if pipeline is not None:

    if not arquivo_nao_vazio(
        pipeline
    ):

        pipeline_valido = False

        validacao_artefatos.append(
            "✗ Pipeline: arquivo vazio."
        )

    else:

        try:

            pipeline_yaml = yaml.safe_load(
                pipeline
            )

            if not isinstance(
                pipeline_yaml,
                dict
            ):
                raise ValueError(
                    "estrutura YAML raiz não é um mapeamento"
                )

            pipeline_valido = True

            validacao_artefatos.append(
                "✓ Pipeline YAML válido."
            )

        except Exception as erro:

            pipeline_valido = False

            validacao_artefatos.append(
                f"✗ Pipeline YAML inválido: {erro}"
            )


# ---------------------------------------------------------
# 2. .csproj XML
# ---------------------------------------------------------

if csproj is not None:

    if not arquivo_nao_vazio(
        csproj
    ):

        csproj_valido = False

        validacao_artefatos.append(
            "✗ .csproj: arquivo vazio."
        )

    else:

        try:

            ET.fromstring(
                csproj
            )

            csproj_valido = True

            validacao_artefatos.append(
                "✓ .csproj XML válido."
            )

        except ET.ParseError as erro:

            csproj_valido = False

            validacao_artefatos.append(
                f"✗ .csproj XML inválido: {erro}"
            )


# ---------------------------------------------------------
# 3. renovate.json
# ---------------------------------------------------------

if renovate_json is not None:

    if not arquivo_nao_vazio(
        renovate_json
    ):

        renovate_json_valido = False

        validacao_artefatos.append(
            "✗ renovate.json: arquivo vazio."
        )

    else:

        try:

            renovate_data_validacao = json.loads(
                renovate_json
            )

            if not isinstance(
                renovate_data_validacao,
                dict
            ):
                raise ValueError(
                    "raiz JSON não é um objeto"
                )

            renovate_json_valido = True

            validacao_artefatos.append(
                "✓ renovate.json válido."
            )

        except Exception as erro:

            renovate_json_valido = False

            validacao_artefatos.append(
                f"✗ renovate.json inválido: {erro}"
            )


# ---------------------------------------------------------
# 4. RENOVATE_SETUP.md
# ---------------------------------------------------------

if renovate_setup is not None:

    if arquivo_nao_vazio(
        renovate_setup
    ):

        renovate_setup_valido = True

        validacao_artefatos.append(
            "✓ RENOVATE_SETUP.md legível e não vazio."
        )

    else:

        renovate_setup_valido = False

        validacao_artefatos.append(
            "✗ RENOVATE_SETUP.md: arquivo vazio."
        )


# ---------------------------------------------------------
# 5. Dockerfile
# ---------------------------------------------------------

if dockerfile is not None:

    if not arquivo_nao_vazio(
        dockerfile
    ):

        dockerfile_valido = False

        validacao_artefatos.append(
            "✗ Dockerfile: arquivo vazio."
        )

    else:

        possui_from = bool(
            re.search(
                r"^\s*FROM\s+\S+",
                dockerfile,
                flags=(
                    re.IGNORECASE
                    | re.MULTILINE
                )
            )
        )

        if possui_from:

            dockerfile_valido = True

            validacao_artefatos.append(
                "✓ Dockerfile estruturalmente válido "
                "(instrução FROM identificada)."
            )

        else:

            dockerfile_valido = False

            validacao_artefatos.append(
                "✗ Dockerfile sem instrução FROM identificada."
            )


# ---------------------------------------------------------
# 6. Docker Compose
# ---------------------------------------------------------

if docker_compose is not None:

    if not arquivo_nao_vazio(
        docker_compose
    ):

        docker_compose_valido = False

        validacao_artefatos.append(
            "✗ Docker Compose: arquivo vazio."
        )

    else:

        try:

            docker_compose_obj = yaml.safe_load(
                docker_compose
            )

            if not isinstance(
                docker_compose_obj,
                dict
            ):
                raise ValueError(
                    "estrutura YAML raiz não é um mapeamento"
                )

            if "services" not in docker_compose_obj:

                docker_compose_valido = False

                validacao_artefatos.append(
                    "✗ Docker Compose YAML válido, "
                    "mas sem seção 'services'."
                )

            else:

                docker_compose_valido = True

                validacao_artefatos.append(
                    "✓ Docker Compose YAML válido "
                    "e seção 'services' identificada."
                )

        except Exception as erro:

            docker_compose_valido = False

            validacao_artefatos.append(
                f"✗ Docker Compose YAML inválido: {erro}"
            )


# ---------------------------------------------------------
# 7. Program.cs
# ---------------------------------------------------------

if program_cs is not None:

    if not arquivo_nao_vazio(
        program_cs
    ):

        program_cs_valido = False

        validacao_artefatos.append(
            "✗ Program.cs: arquivo vazio."
        )

    else:

        indicadores_cs = (
            "WebApplication",
            "CreateBuilder",
            "builder.",
            "Build("
        )

        program_cs_valido = any(
            indicador.lower()
            in program_cs.lower()
            for indicador in indicadores_cs
        )

        if program_cs_valido:

            validacao_artefatos.append(
                "✓ Program.cs legível e com estrutura "
                "compatível com inicialização de aplicação .NET."
            )

        else:

            validacao_artefatos.append(
                "⚠ Program.cs legível, mas a estrutura "
                "esperada de inicialização .NET "
                "não foi identificada."
            )


# ---------------------------------------------------------
# 8. appsettings.json
# ---------------------------------------------------------

if appsettings is not None:

    if not arquivo_nao_vazio(
        appsettings
    ):

        appsettings_valido = False

        validacao_artefatos.append(
            "✗ appsettings.json: arquivo vazio."
        )

    else:

        try:

            appsettings_obj = json.loads(
                appsettings
            )

            if not isinstance(
                appsettings_obj,
                dict
            ):
                raise ValueError(
                    "raiz JSON não é um objeto"
                )

            appsettings_valido = True

            validacao_artefatos.append(
                "✓ appsettings.json válido."
            )

        except Exception as erro:

            appsettings_valido = False

            validacao_artefatos.append(
                f"✗ appsettings.json inválido: {erro}"
            )


# ---------------------------------------------------------
# 9. .editorconfig
# ---------------------------------------------------------

if editorconfig is not None:

    if not arquivo_nao_vazio(
        editorconfig
    ):

        editorconfig_valido = False

        validacao_artefatos.append(
            "✗ .editorconfig: arquivo vazio."
        )

    else:

        possui_secao = bool(
            re.search(
                r"^\s*\[[^\]]+\]\s*$",
                editorconfig,
                flags=re.MULTILINE
            )
        )

        editorconfig_valido = possui_secao

        if possui_secao:

            validacao_artefatos.append(
                "✓ .editorconfig estruturalmente válido "
                "(seção de configuração identificada)."
            )

        else:

            validacao_artefatos.append(
                "⚠ .editorconfig legível, "
                "mas nenhuma seção foi identificada."
            )


# ---------------------------------------------------------
# 10. Arquivos de ambiente
# ---------------------------------------------------------

env_example_valido = validar_env(
    env_example,
    ".env.example"
)

env_io_example_valido = validar_env(
    env_io_example,
    ".env.io.example"
)


# ---------------------------------------------------------
# 11. Scripts
# ---------------------------------------------------------

for script_nome in sorted(
    scripts_nomes
):

    conteudo_script = scripts.get(
        script_nome
    )

    if arquivo_nao_vazio(
        conteudo_script
    ):

        scripts_validos[
            script_nome
        ] = True

        extensao = Path(
            normalizar_nome(
                script_nome
            )
        ).suffix

        tipo_script = {
            ".sql": "SQL",
            ".ps1": "PowerShell",
            ".sh": "Shell",
            ".bat": "Batch",
            ".cmd": "Command"
        }.get(
            extensao,
            "script"
        )

        validacao_artefatos.append(
            f"✓ {script_nome}: "
            f"script {tipo_script} legível e não vazio."
        )

    else:

        scripts_validos[
            script_nome
        ] = False

        validacao_artefatos.append(
            f"✗ {script_nome}: arquivo vazio."
        )


# ---------------------------------------------------------
# Resultado da identificação
# ---------------------------------------------------------

print(
    "ARTEFATOS IDENTIFICADOS\n"
)


def imprimir_artefato(
    descricao,
    nome
):
    if nome:

        print(
            f"✓ {descricao}: {nome}"
        )

    else:

        print(
            f"? {descricao}: não fornecido"
        )


imprimir_artefato(
    "Pipeline",
    pipeline_nome
)

imprimir_artefato(
    ".csproj",
    csproj_nome
)

imprimir_artefato(
    "renovate.json",
    renovate_json_nome
)

imprimir_artefato(
    "RENOVATE_SETUP.md",
    renovate_setup_nome
)

imprimir_artefato(
    "Dockerfile",
    dockerfile_nome
)

imprimir_artefato(
    "Docker Compose",
    docker_compose_nome
)

imprimir_artefato(
    "Program.cs",
    program_cs_nome
)

imprimir_artefato(
    "appsettings.json",
    appsettings_nome
)

imprimir_artefato(
    ".editorconfig",
    editorconfig_nome
)

imprimir_artefato(
    ".env.example",
    env_example_nome
)

imprimir_artefato(
    ".env.io.example",
    env_io_example_nome
)


if scripts_nomes:

    print(
        f"✓ Scripts: "
        f"{len(scripts_nomes)} arquivo(s)"
    )

    for nome in sorted(
        scripts_nomes
    ):
        print(
            f"    - {nome}"
        )

else:

    print(
        "? Scripts: não fornecidos"
    )


# ---------------------------------------------------------
# Resultado da validação
# ---------------------------------------------------------

print(
    "\nVALIDAÇÃO\n"
)

if validacao_artefatos:

    for mensagem in validacao_artefatos:

        print(
            mensagem
        )

else:

    print(
        "Nenhum artefato sujeito a validação "
        "foi fornecido."
    )


# ---------------------------------------------------------
# Estado técnico da validação
# ---------------------------------------------------------

print(
    "\nESTADO DA VALIDAÇÃO\n"
)

estados_validacao = {
    "Pipeline": pipeline_valido,
    ".csproj": csproj_valido,
    "renovate.json": renovate_json_valido,
    "RENOVATE_SETUP.md": renovate_setup_valido,
    "Dockerfile": dockerfile_valido,
    "Docker Compose": docker_compose_valido,
    "Program.cs": program_cs_valido,
    "appsettings.json": appsettings_valido,
    ".editorconfig": editorconfig_valido,
    ".env.example": env_example_valido,
    ".env.io.example": env_io_example_valido
}


for artefato, estado in estados_validacao.items():

    if estado is True:
        simbolo = "✓"

    elif estado is False:
        simbolo = "✗"

    else:
        simbolo = "?"

    print(
        f"{simbolo} {artefato}"
    )
```

```python
# ---------------------------------------------------------
# PIPELINE MATURITY CLASSIFICATION
# ---------------------------------------------------------

def classificar_maturidade(e):

    if not e:

        return {
            "nivel": None,
            "classificacao": "Indeterminado",
            "motivo": (
                "Nenhuma evidência foi extraída."
            )
        }


    gitlab_ci = e.get(
        "GitLab CI como código"
    )

    restore = e.get(
        "`dotnet restore` no CI"
    )

    build = e.get(
        "Build Release no CI"
    )

    sonar = e.get(
        "SonarQube"
    )

    quality_gate_wait = e.get(
        "Espera pelo Quality Gate"
    )

    quality_gate_blocking = e.get(
        "Quality Gate bloqueante"
    )

    testes = e.get(
        "Testes automatizados"
    )

    deploy = e.get(
        "Deploy automatizado"
    )

    promocao = e.get(
        "Promoção automática"
    )

    rollback = e.get(
        "Rollback automatizado"
    )


    # -----------------------------------------------------
    # Evidência insuficiente
    # -----------------------------------------------------

    if all(
        valor is None
        for valor in (
            gitlab_ci,
            restore,
            build,
            sonar
        )
    ):

        return {
            "nivel": None,
            "classificacao": "Indeterminado",
            "motivo": (
                "O pipeline não está disponível ou não pôde "
                "ser validado."
            )
        }


    # -----------------------------------------------------
    # CI estruturada
    # -----------------------------------------------------

    ci_estruturada = (
        gitlab_ci is True
        and restore is True
        and build is True
    )

    qualidade = (
        sonar is True
        or quality_gate_wait is True
    )


    # -----------------------------------------------------
    # Nível 4
    # -----------------------------------------------------

    if (
        ci_estruturada
        and qualidade
        and testes is True
        and quality_gate_blocking is True
        and deploy is True
        and promocao is True
        and rollback is True
    ):

        return {
            "nivel": 4,
            "classificacao": "Avançado/Consolidado",
            "motivo": (
                "O pipeline apresenta CI estruturada, testes "
                "automatizados, controle de qualidade bloqueante, "
                "deploy, promoção e rollback automatizados."
            )
        }


    # -----------------------------------------------------
    # Nível 3
    # -----------------------------------------------------

    if (
        ci_estruturada
        and qualidade
    ):

        return {
            "nivel": 3,
            "classificacao": "Parcial/Emergente",
            "motivo": (
                "O pipeline apresenta integração contínua "
                "estruturada e controles de qualidade, mas "
                "a automação de ponta a ponta permanece incompleta."
            )
        }


    # -----------------------------------------------------
    # Nível 2
    # -----------------------------------------------------

    if (
        gitlab_ci is True
        and (
            restore is True
            or build is True
            or sonar is True
        )
    ):

        return {
            "nivel": 2,
            "classificacao": "Inicial/Ad hoc",
            "motivo": (
                "Existe pipeline versionado com automações "
                "pontuais, mas sem CI estruturada completa."
            )
        }


    # -----------------------------------------------------
    # Nível 1
    # -----------------------------------------------------

    if gitlab_ci is True:

        return {
            "nivel": 1,
            "classificacao": "Ausente/Tradicional",
            "motivo": (
                "O pipeline existe, mas não foram identificadas "
                "automações relevantes segundo os critérios definidos."
            )
        }


    return {
        "nivel": None,
        "classificacao": "Indeterminado",
        "motivo": (
            "As evidências disponíveis são insuficientes."
        )
    }


# ---------------------------------------------------------
# Executa a classificação
# ---------------------------------------------------------

resultado_maturidade = classificar_maturidade(
    evidencias
)


# ---------------------------------------------------------
# Resultado
# ---------------------------------------------------------

print("CLASSIFICAÇÃO DE MATURIDADE\n")

print(
    "Nível:",
    resultado_maturidade["nivel"]
)

print(
    "Classificação:",
    resultado_maturidade["classificacao"]
)

print(
    "Motivo:",
    resultado_maturidade["motivo"]
)
```

```python
# ---------------------------------------------------------
# AUTOMATED EVIDENCE EXTRACTION
# ---------------------------------------------------------

target_framework = None
sdk_pipeline = None

# Reinicia as evidências a cada execução
evidencias = {}


# ---------------------------------------------------------
# 1. .csproj
# ---------------------------------------------------------

if csproj:

    try:
        root = ET.fromstring(csproj)

        for elemento in root.iter():

            tag = elemento.tag.split("}")[-1]

            if tag == "TargetFramework":

                target_framework = (
                    elemento.text.strip()
                    if elemento.text
                    else None
                )

                break

            elif tag == "TargetFrameworks":

                target_framework = (
                    elemento.text.strip()
                    if elemento.text
                    else None
                )

                break

    except ET.ParseError as erro:

        print(
            "⚠ Não foi possível interpretar o arquivo .csproj:"
        )

        print(erro)


# ---------------------------------------------------------
# 2. GitLab CI
# ---------------------------------------------------------

if pipeline:

    # SDK .NET utilizado no pipeline
    match_sdk = buscar_regex(
        pipeline,
        r"mcr\.microsoft\.com/dotnet/sdk:([0-9.]+)"
    )

    if match_sdk:
        sdk_pipeline = match_sdk.group(1)


    # Pipeline versionado
    evidencias["GitLab CI como código"] = True


    # Restore
    evidencias["`dotnet restore` no CI"] = contem(
        pipeline,
        "dotnet restore"
    )


    # Build Release
    possui_build = contem(
        pipeline,
        "dotnet build"
    )

    possui_release = contem(
        pipeline,
        "--configuration release",
        "-c release"
    )

    evidencias["Build Release no CI"] = (
        possui_build
        and possui_release
    )


    # SonarQube
    evidencias["SonarQube"] = contem(
        pipeline,
        "dotnet-sonarscanner",
        "sonarscanner",
        "sonar.host.url",
        "sonar_token"
    )


    # Quality Gate
    evidencias["Espera pelo Quality Gate"] = contem(
        pipeline,
        "sonar.qualitygate.wait=true"
    )


    # Quality Gate bloqueante
    quality_gate_wait = evidencias[
        "Espera pelo Quality Gate"
    ]

    allow_failure_match = buscar_regex(
        pipeline,
        r"allow_failure\s*:\s*true"
    )

    allow_failure_true = (
        allow_failure_match is not None
    )

    evidencias["Quality Gate bloqueante"] = (
        quality_gate_wait
        and not allow_failure_true
    )


    # Cache Sonar
    evidencias["Cache Sonar"] = contem(
        pipeline,
        "${sonar_user_home}/cache",
        ".sonar"
    )


    # Cache NuGet
    evidencias["Cache NuGet"] = contem(
        pipeline,
        ".nuget/packages"
    )


    # Merge Request
    evidencias["Regra para Merge Request"] = contem(
        pipeline,
        "merge_request_event"
    )


    # Branch main
    possui_variavel_branch = contem(
        pipeline,
        "ci_commit_branch"
    )

    possui_main_match = buscar_regex(
        pipeline,
        r"ci_commit_branch.*main"
    )

    possui_main = (
        possui_main_match is not None
    )

    evidencias["Regra para branch main"] = (
        possui_variavel_branch
        and possui_main
    )


    # Testes automatizados
    evidencias["Testes automatizados"] = contem(
        pipeline,
        "dotnet test",
        "pytest",
        "npm test",
        "npm run test",
        "mvn test",
        "gradle test"
    )


    # Deploy automatizado
    evidencias["Deploy automatizado"] = contem(
        pipeline,
        "kubectl apply",
        "helm upgrade",
        "docker stack deploy",
        "docker compose up",
        "docker-compose up",
        "oc apply"
    )


    # Promoção automática
    evidencias["Promoção automática"] = contem(
        pipeline,
        "promote",
        "promotion",
        "promote-to-production",
        "promote_to_production"
    )


    # Rollback automatizado
    evidencias["Rollback automatizado"] = contem(
        pipeline,
        "helm rollback",
        "kubectl rollout undo",
        "docker service rollback"
    )

else:

    evidencias["GitLab CI como código"] = None
    evidencias["`dotnet restore` no CI"] = None
    evidencias["Build Release no CI"] = None
    evidencias["SonarQube"] = None
    evidencias["Espera pelo Quality Gate"] = None
    evidencias["Quality Gate bloqueante"] = None
    evidencias["Cache Sonar"] = None
    evidencias["Cache NuGet"] = None
    evidencias["Regra para Merge Request"] = None
    evidencias["Regra para branch main"] = None
    evidencias["Testes automatizados"] = None
    evidencias["Deploy automatizado"] = None
    evidencias["Promoção automática"] = None
    evidencias["Rollback automatizado"] = None


# ---------------------------------------------------------
# 3. Renovate
# ---------------------------------------------------------

if renovate_json:

    evidencias["Renovate configurado"] = True

    try:

        renovate_data = json.loads(
            renovate_json
        )

        enabled_managers = renovate_data.get(
            "enabledManagers",
            []
        )

        evidencias[
            "Gerenciamento NuGet pelo Renovate"
        ] = (
            "nuget" in enabled_managers
        )

        evidencias[
            "Gerenciamento Dockerfile pelo Renovate"
        ] = (
            "dockerfile" in enabled_managers
        )


        # Vulnerability Alerts
        vulnerability_alerts = renovate_data.get(
            "vulnerabilityAlerts"
        )

        if vulnerability_alerts is not None:

            evidencias[
                "Vulnerability Alerts"
            ] = True

        else:

            evidencias[
                "Vulnerability Alerts"
            ] = contem(
                renovate_json,
                "vulnerabilityalerts",
                "osv"
            )


        # Lock File Maintenance
        lock_file_maintenance = renovate_data.get(
            "lockFileMaintenance"
        )

        if lock_file_maintenance is not None:

            evidencias[
                "Lock File Maintenance"
            ] = True

        else:

            evidencias[
                "Lock File Maintenance"
            ] = contem(
                renovate_json,
                "lockfilemaintenance"
            )


    except json.JSONDecodeError:

        print(
            "⚠ renovate.json inválido. "
            "A análise será feita por busca textual."
        )

        evidencias[
            "Gerenciamento NuGet pelo Renovate"
        ] = contem(
            renovate_json,
            '"nuget"'
        )

        evidencias[
            "Gerenciamento Dockerfile pelo Renovate"
        ] = contem(
            renovate_json,
            '"dockerfile"'
        )

        evidencias[
            "Vulnerability Alerts"
        ] = contem(
            renovate_json,
            "vulnerabilityalerts",
            "osv"
        )

        evidencias[
            "Lock File Maintenance"
        ] = contem(
            renovate_json,
            "lockfilemaintenance"
        )

else:

    evidencias["Renovate configurado"] = None

    evidencias[
        "Gerenciamento NuGet pelo Renovate"
    ] = None

    evidencias[
        "Gerenciamento Dockerfile pelo Renovate"
    ] = None

    evidencias[
        "Vulnerability Alerts"
    ] = None

    evidencias[
        "Lock File Maintenance"
    ] = None


# ---------------------------------------------------------
# 4. Renovate orquestrado no pipeline
# ---------------------------------------------------------

if pipeline:

    possui_imagem_renovate = contem(
        pipeline,
        "renovate/renovate"
    )

    possui_comando_renovate = contem(
        pipeline,
        "renovate $ci_project_path",
        "renovate ${ci_project_path}"
    )

    evidencias[
        "Job Renovate no pipeline"
    ] = (
        possui_imagem_renovate
        or possui_comando_renovate
    )


    evidencias[
        "Regra schedule para Renovate"
    ] = (
        evidencias[
            "Job Renovate no pipeline"
        ]
        and contem(
            pipeline,
            "schedule"
        )
    )


    evidencias[
        "Execução manual do Renovate"
    ] = (
        evidencias[
            "Job Renovate no pipeline"
        ]
        and contem(
            pipeline,
            "when: manual",
            "ci_pipeline_source == \"web\"",
            "ci_pipeline_source == 'web'"
        )
    )

else:

    evidencias[
        "Job Renovate no pipeline"
    ] = None

    evidencias[
        "Regra schedule para Renovate"
    ] = None

    evidencias[
        "Execução manual do Renovate"
    ] = None


# Configuração e job não comprovam execução efetiva
evidencias[
    "Execução efetiva do Renovate"
] = None


# ---------------------------------------------------------
# 5. Dockerfile
# ---------------------------------------------------------

if dockerfile:

    evidencias[
        "Dockerfile presente"
    ] = True

    # Detecta múltiplos FROM
    quantidade_from = len(
        re.findall(
            r"^\s*FROM\s+",
            dockerfile,
            flags=re.IGNORECASE | re.MULTILINE
        )
    )

    evidencias[
        "Docker multi-stage build"
    ] = (
        quantidade_from >= 2
    )

    evidencias[
        "Restore no Dockerfile"
    ] = contem(
        dockerfile,
        "dotnet restore"
    )

    evidencias[
        "Build no Dockerfile"
    ] = contem(
        dockerfile,
        "dotnet build"
    )

    evidencias[
        "Publish no Dockerfile"
    ] = contem(
        dockerfile,
        "dotnet publish"
    )

else:

    evidencias[
        "Dockerfile presente"
    ] = None

    evidencias[
        "Docker multi-stage build"
    ] = None

    evidencias[
        "Restore no Dockerfile"
    ] = None

    evidencias[
        "Build no Dockerfile"
    ] = None

    evidencias[
        "Publish no Dockerfile"
    ] = None


# ---------------------------------------------------------
# 6. Docker Compose
# ---------------------------------------------------------

if docker_compose:

    evidencias[
        "Docker Compose presente"
    ] = True

    evidencias[
        "Build definido no Docker Compose"
    ] = contem(
        docker_compose,
        "build:"
    )

    evidencias[
        "Política de restart"
    ] = contem(
        docker_compose,
        "restart:"
    )

    evidencias[
        "Rede externa no Docker Compose"
    ] = (
        contem(
            docker_compose,
            "external:"
        )
        and contem(
            docker_compose,
            "true"
        )
    )

    evidencias[
        "Reverse proxy configurado"
    ] = contem(
        docker_compose,
        "traefik",
        "nginx",
        "reverse proxy"
    )

else:

    evidencias[
        "Docker Compose presente"
    ] = None

    evidencias[
        "Build definido no Docker Compose"
    ] = None

    evidencias[
        "Política de restart"
    ] = None

    evidencias[
        "Rede externa no Docker Compose"
    ] = None

    evidencias[
        "Reverse proxy configurado"
    ] = None


# ---------------------------------------------------------
# 7. Program.cs / Observabilidade
# ---------------------------------------------------------

if program_cs:

    evidencias[
        "Sentry integrado em runtime"
    ] = contem(
        program_cs,
        "usesentry",
        "sentry"
    )

    evidencias[
        "Health Checks"
    ] = contem(
        program_cs,
        "addhealthchecks",
        "maphealthchecks"
    )

    evidencias[
        "Health Check de banco"
    ] = contem(
        program_cs,
        "addnpgsql",
        "npgsql",
        "healthcheck"
    )

    evidencias[
        "Health Check de API"
    ] = (
        contem(
            program_cs,
            "addurlgroup",
            "httpclient",
            "healthcheck"
        )
        and contem(
            program_cs,
            "health"
        )
    )

    evidencias[
        "Console Logging"
    ] = contem(
        program_cs,
        "addconsole",
        "console"
    )

else:

    evidencias[
        "Sentry integrado em runtime"
    ] = None

    evidencias[
        "Health Checks"
    ] = None

    evidencias[
        "Health Check de banco"
    ] = None

    evidencias[
        "Health Check de API"
    ] = None

    evidencias[
        "Console Logging"
    ] = None


# ---------------------------------------------------------
# 8. appsettings
# ---------------------------------------------------------

if appsettings:

    evidencias[
        "appsettings presente"
    ] = True

    evidencias[
        "Sentry configurado"
    ] = contem(
        appsettings,
        "sentry",
        "dsn"
    )

    evidencias[
        "Matomo configurado"
    ] = contem(
        appsettings,
        "matomo"
    )

    evidencias[
        "PostgreSQL configurado"
    ] = contem(
        appsettings,
        "postgres",
        "npgsql"
    )

    evidencias[
        "Configuração externalizada"
    ] = contem(
        appsettings,
        "environment",
        "connectionstrings",
        "configuration"
    )

else:

    evidencias[
        "appsettings presente"
    ] = None

    evidencias[
        "Sentry configurado"
    ] = None

    evidencias[
        "Matomo configurado"
    ] = None

    evidencias[
        "PostgreSQL configurado"
    ] = None

    evidencias[
        "Configuração externalizada"
    ] = None


# ---------------------------------------------------------
# 9. .editorconfig
# ---------------------------------------------------------

if editorconfig:

    evidencias[
        ".editorconfig presente"
    ] = True

    evidencias[
        "Regras de analyzer declaradas"
    ] = contem(
        editorconfig,
        "dotnet_diagnostic",
        "severity"
    )

    evidencias[
        "Regras de segurança declaradas"
    ] = contem(
        editorconfig,
        "security",
        "ca2",
        "s"
    )

else:

    evidencias[
        ".editorconfig presente"
    ] = None

    evidencias[
        "Regras de analyzer declaradas"
    ] = None

    evidencias[
        "Regras de segurança declaradas"
    ] = None


# ---------------------------------------------------------
# 10. .env.example
# ---------------------------------------------------------

if env_example:

    evidencias[
        ".env.example presente"
    ] = True

    evidencias[
        "Parametrização de ambientes"
    ] = contem(
        env_example,
        "environment",
        "env",
        "stage",
        "server",
        "host"
    )

else:

    evidencias[
        ".env.example presente"
    ] = None

    evidencias[
        "Parametrização de ambientes"
    ] = None


# ---------------------------------------------------------
# 11. .env.io.example
# ---------------------------------------------------------

if env_io_example:

    evidencias[
        ".env.io.example presente"
    ] = True

    # Se já havia evidência anterior, preserva True
    evidencia_parametrizacao = contem(
        env_io_example,
        "environment",
        "stage",
        "server",
        "version",
        "deployer",
        "project"
    )

    valor_atual = evidencias.get(
        "Parametrização de ambientes"
    )

    if valor_atual is True:
        evidencias[
            "Parametrização de ambientes"
        ] = True

    elif valor_atual is False:
        evidencias[
            "Parametrização de ambientes"
        ] = evidencia_parametrizacao

    else:
        evidencias[
            "Parametrização de ambientes"
        ] = evidencia_parametrizacao

else:

    evidencias[
        ".env.io.example presente"
    ] = None


# ---------------------------------------------------------
# 12. Scripts
# ---------------------------------------------------------

if scripts:

    evidencias[
        "Scripts presentes"
    ] = True

    conteudo_scripts = "\n".join(
        str(conteudo)
        for conteudo in scripts.values()
    )

    nomes_scripts = "\n".join(
        str(nome)
        for nome in scripts.keys()
    )

    conteudo_total_scripts = (
        nomes_scripts
        + "\n"
        + conteudo_scripts
    )

    evidencias[
        "Script de inicialização de banco"
    ] = contem(
        conteudo_total_scripts,
        "init.sql",
        "create table",
        "create schema"
    )

    evidencias[
        "Script de seed"
    ] = contem(
        conteudo_total_scripts,
        "seed.sql",
        "insert into"
    )

    evidencias[
        "Dump de banco presente"
    ] = contem(
        conteudo_total_scripts,
        "pg_dump",
        ".dump",
        "postgresql database dump"
    )

    evidencias[
        "Automação de ambiente local"
    ] = contem(
        conteudo_total_scripts,
        "docker compose",
        "docker-compose",
        "start-db",
        "init.sql",
        "localhost"
    )

else:

    evidencias[
        "Scripts presentes"
    ] = None

    evidencias[
        "Script de inicialização de banco"
    ] = None

    evidencias[
        "Script de seed"
    ] = None

    evidencias[
        "Dump de banco presente"
    ] = None

    evidencias[
        "Automação de ambiente local"
    ] = None


# ---------------------------------------------------------
# 13. Resumo tecnológico
# ---------------------------------------------------------

tecnologia = {
    "target_framework": target_framework,
    "sdk_pipeline": sdk_pipeline
}


# ---------------------------------------------------------
# 14. Resultado
# ---------------------------------------------------------

print("EVIDÊNCIAS EXTRAÍDAS\n")

print(
    "Target Framework:",
    target_framework
)

print(
    "SDK Pipeline:",
    sdk_pipeline
)

print()


# ---------------------------------------------------------
# 15. Exibição tri-state
# ---------------------------------------------------------

for nome, valor in evidencias.items():

    if valor is True:
        simbolo = "✓"

    elif valor is False:
        simbolo = "—"

    else:
        simbolo = "?"

    print(
        f"{simbolo} {nome}"
    )
```

```python
# ---------------------------------------------------------
# ARTIFACT IDENTIFICATION AND VALIDATION
# ---------------------------------------------------------

if not arquivos:
    raise ValueError(
        "Nenhum arquivo foi carregado. "
        "Execute primeiro a etapa Artifact Upload."
    )


# ---------------------------------------------------------
# Reinicia variáveis de identificação
# ---------------------------------------------------------

pipeline = None
pipeline_nome = None

csproj = None
csproj_nome = None

renovate_json = None
renovate_json_nome = None

renovate_setup = None
renovate_setup_nome = None

dockerfile = None
dockerfile_nome = None

docker_compose = None
docker_compose_nome = None

program_cs = None
program_cs_nome = None

appsettings = None
appsettings_nome = None

editorconfig = None
editorconfig_nome = None

env_example = None
env_example_nome = None

env_io_example = None
env_io_example_nome = None

scripts = {}
scripts_nomes = []

validacao_artefatos = []


# ---------------------------------------------------------
# Estados formais de validação
#
# None  = artefato não fornecido / não validável
# True  = artefato fornecido e validado
# False = artefato fornecido, porém inválido
# ---------------------------------------------------------

pipeline_valido = None
csproj_valido = None
renovate_json_valido = None
renovate_setup_valido = None
dockerfile_valido = None
docker_compose_valido = None
program_cs_valido = None
appsettings_valido = None
editorconfig_valido = None
env_example_valido = None
env_io_example_valido = None

scripts_validos = {}


# ---------------------------------------------------------
# Normalização de nomes
# ---------------------------------------------------------

def nome_base(nome):
    return (
        nome
        .replace("\\", "/")
        .split("/")[-1]
        .strip()
    )


def normalizar_nome(nome):
    nome = nome_base(nome)

    # Remove sufixos adicionados pelo Colab:
    # arquivo (1).json -> arquivo.json
    # Dockerfile (2)   -> Dockerfile
    nome = re.sub(
        r"\s*\(\d+\)(?=\.|$)",
        "",
        nome
    )

    return nome.lower().strip()


# ---------------------------------------------------------
# Seleção de artefato único
# ---------------------------------------------------------

def selecionar_unico(
    candidatos,
    descricao
):
    if not candidatos:
        return None, None

    if len(candidatos) > 1:

        nomes = ", ".join(
            sorted(
                nome
                for nome, _ in candidatos
            )
        )

        raise ValueError(
            f"Mais de um arquivo foi identificado como "
            f"{descricao}: {nomes}. "
            "Forneça apenas um artefato desse tipo por versão."
        )

    return (
        candidatos[0][1],
        candidatos[0][0]
    )


# ---------------------------------------------------------
# Conversão segura para texto
# ---------------------------------------------------------

def como_texto(conteudo):
    if conteudo is None:
        return None

    if isinstance(
        conteudo,
        bytes
    ):
        return conteudo.decode(
            "utf-8-sig",
            errors="replace"
        )

    texto = str(
        conteudo
    )

    # Remove BOM caso o conteúdo já seja string.
    return texto.lstrip(
        "\ufeff"
    )


# ---------------------------------------------------------
# Candidatos
# ---------------------------------------------------------

candidatos_pipeline = []
candidatos_csproj = []
candidatos_renovate = []
candidatos_renovate_setup = []
candidatos_dockerfile = []
candidatos_docker_compose = []
candidatos_program = []
candidatos_appsettings = []
candidatos_editorconfig = []
candidatos_env_example = []
candidatos_env_io_example = []

extensoes_script = {
    ".sql",
    ".ps1",
    ".sh",
    ".bat",
    ".cmd"
}


# ---------------------------------------------------------
# Identificação dos arquivos
# ---------------------------------------------------------

for nome, conteudo in arquivos.items():

    base = nome_base(
        nome
    )

    normalizado = normalizar_nome(
        nome
    )


    # GitLab CI
    if normalizado in {
        ".gitlab-ci.yml",
        ".gitlab-ci.yaml",
        "gitlab-ci.yml",
        "gitlab-ci.yaml"
    }:

        candidatos_pipeline.append(
            (
                base,
                conteudo
            )
        )

        continue


    # Projeto .NET
    if normalizado.endswith(
        ".csproj"
    ):

        candidatos_csproj.append(
            (
                base,
                conteudo
            )
        )

        continue


    # Renovate
    if normalizado == "renovate.json":

        candidatos_renovate.append(
            (
                base,
                conteudo
            )
        )

        continue


    if normalizado == "renovate_setup.md":

        candidatos_renovate_setup.append(
            (
                base,
                conteudo
            )
        )

        continue


    # Dockerfile
    if normalizado == "dockerfile":

        candidatos_dockerfile.append(
            (
                base,
                conteudo
            )
        )

        continue


    # Docker Compose
    if normalizado in {
        "docker-compose.yml",
        "docker-compose.yaml",
        "compose.yml",
        "compose.yaml"
    }:

        candidatos_docker_compose.append(
            (
                base,
                conteudo
            )
        )

        continue


    # Program.cs
    if normalizado == "program.cs":

        candidatos_program.append(
            (
                base,
                conteudo
            )
        )

        continue


    # appsettings.json
    if normalizado == "appsettings.json":

        candidatos_appsettings.append(
            (
                base,
                conteudo
            )
        )

        continue


    # .editorconfig
    if normalizado == ".editorconfig":

        candidatos_editorconfig.append(
            (
                base,
                conteudo
            )
        )

        continue


    # .env.io.example
    if normalizado in {
        ".env.io.example",
        "env.io.example"
    }:

        candidatos_env_io_example.append(
            (
                base,
                conteudo
            )
        )

        continue


    # .env.example
    if normalizado in {
        ".env.example",
        "env.example"
    }:

        candidatos_env_example.append(
            (
                base,
                conteudo
            )
        )

        continue


    # Scripts
    extensao = Path(
        normalizado
    ).suffix

    if extensao in extensoes_script:

        scripts[
            base
        ] = como_texto(
            conteudo
        )

        scripts_nomes.append(
            base
        )


# ---------------------------------------------------------
# Seleção dos artefatos
# ---------------------------------------------------------

pipeline, pipeline_nome = selecionar_unico(
    candidatos_pipeline,
    "pipeline GitLab CI"
)

csproj, csproj_nome = selecionar_unico(
    candidatos_csproj,
    "arquivo .csproj"
)

renovate_json, renovate_json_nome = selecionar_unico(
    candidatos_renovate,
    "renovate.json"
)

renovate_setup, renovate_setup_nome = selecionar_unico(
    candidatos_renovate_setup,
    "RENOVATE_SETUP.md"
)

dockerfile, dockerfile_nome = selecionar_unico(
    candidatos_dockerfile,
    "Dockerfile"
)

docker_compose, docker_compose_nome = selecionar_unico(
    candidatos_docker_compose,
    "Docker Compose"
)

program_cs, program_cs_nome = selecionar_unico(
    candidatos_program,
    "Program.cs"
)

appsettings, appsettings_nome = selecionar_unico(
    candidatos_appsettings,
    "appsettings.json"
)

editorconfig, editorconfig_nome = selecionar_unico(
    candidatos_editorconfig,
    ".editorconfig"
)

env_example, env_example_nome = selecionar_unico(
    candidatos_env_example,
    ".env.example"
)

env_io_example, env_io_example_nome = selecionar_unico(
    candidatos_env_io_example,
    ".env.io.example"
)


# ---------------------------------------------------------
# Conversão para texto
# ---------------------------------------------------------

pipeline = como_texto(
    pipeline
)

csproj = como_texto(
    csproj
)

renovate_json = como_texto(
    renovate_json
)

renovate_setup = como_texto(
    renovate_setup
)

dockerfile = como_texto(
    dockerfile
)

docker_compose = como_texto(
    docker_compose
)

program_cs = como_texto(
    program_cs
)

appsettings = como_texto(
    appsettings
)

editorconfig = como_texto(
    editorconfig
)

env_example = como_texto(
    env_example
)

env_io_example = como_texto(
    env_io_example
)


# ---------------------------------------------------------
# Funções auxiliares de validação
# ---------------------------------------------------------

def arquivo_nao_vazio(
    conteudo
):
    return (
        conteudo is not None
        and bool(
            conteudo.strip()
        )
    )


def validar_env(
    conteudo,
    descricao
):
    if conteudo is None:
        return None

    if not arquivo_nao_vazio(
        conteudo
    ):
        validacao_artefatos.append(
            f"✗ {descricao}: arquivo vazio."
        )

        return False

    linhas_invalidas = []

    for numero, linha in enumerate(
        conteudo.splitlines(),
        start=1
    ):

        linha = linha.strip()

        if (
            not linha
            or linha.startswith("#")
        ):
            continue

        if "=" not in linha:
            linhas_invalidas.append(
                numero
            )

    if linhas_invalidas:

        validacao_artefatos.append(
            f"⚠ {descricao}: "
            f"{len(linhas_invalidas)} linha(s) "
            "sem formato CHAVE=VALOR."
        )

        # O arquivo continua legível e utilizável
        # para extração de evidências.
        return True

    validacao_artefatos.append(
        f"✓ {descricao} estruturalmente válido."
    )

    return True


# ---------------------------------------------------------
# 1. Pipeline YAML
# ---------------------------------------------------------

if pipeline is not None:

    if not arquivo_nao_vazio(
        pipeline
    ):

        pipeline_valido = False

        validacao_artefatos.append(
            "✗ Pipeline: arquivo vazio."
        )

    else:

        try:

            pipeline_yaml = yaml.safe_load(
                pipeline
            )

            if not isinstance(
                pipeline_yaml,
                dict
            ):
                raise ValueError(
                    "estrutura YAML raiz não é um mapeamento"
                )

            pipeline_valido = True

            validacao_artefatos.append(
                "✓ Pipeline YAML válido."
            )

        except Exception as erro:

            pipeline_valido = False

            validacao_artefatos.append(
                f"✗ Pipeline YAML inválido: {erro}"
            )


# ---------------------------------------------------------
# 2. .csproj XML
# ---------------------------------------------------------

if csproj is not None:

    if not arquivo_nao_vazio(
        csproj
    ):

        csproj_valido = False

        validacao_artefatos.append(
            "✗ .csproj: arquivo vazio."
        )

    else:

        try:

            ET.fromstring(
                csproj
            )

            csproj_valido = True

            validacao_artefatos.append(
                "✓ .csproj XML válido."
            )

        except ET.ParseError as erro:

            csproj_valido = False

            validacao_artefatos.append(
                f"✗ .csproj XML inválido: {erro}"
            )


# ---------------------------------------------------------
# 3. renovate.json
# ---------------------------------------------------------

if renovate_json is not None:

    if not arquivo_nao_vazio(
        renovate_json
    ):

        renovate_json_valido = False

        validacao_artefatos.append(
            "✗ renovate.json: arquivo vazio."
        )

    else:

        try:

            renovate_data_validacao = json.loads(
                renovate_json
            )

            if not isinstance(
                renovate_data_validacao,
                dict
            ):
                raise ValueError(
                    "raiz JSON não é um objeto"
                )

            renovate_json_valido = True

            validacao_artefatos.append(
                "✓ renovate.json válido."
            )

        except Exception as erro:

            renovate_json_valido = False

            validacao_artefatos.append(
                f"✗ renovate.json inválido: {erro}"
            )


# ---------------------------------------------------------
# 4. RENOVATE_SETUP.md
# ---------------------------------------------------------

if renovate_setup is not None:

    if arquivo_nao_vazio(
        renovate_setup
    ):

        renovate_setup_valido = True

        validacao_artefatos.append(
            "✓ RENOVATE_SETUP.md legível e não vazio."
        )

    else:

        renovate_setup_valido = False

        validacao_artefatos.append(
            "✗ RENOVATE_SETUP.md: arquivo vazio."
        )


# ---------------------------------------------------------
# 5. Dockerfile
# ---------------------------------------------------------

if dockerfile is not None:

    if not arquivo_nao_vazio(
        dockerfile
    ):

        dockerfile_valido = False

        validacao_artefatos.append(
            "✗ Dockerfile: arquivo vazio."
        )

    else:

        possui_from = bool(
            re.search(
                r"^\s*FROM\s+\S+",
                dockerfile,
                flags=(
                    re.IGNORECASE
                    | re.MULTILINE
                )
            )
        )

        if possui_from:

            dockerfile_valido = True

            validacao_artefatos.append(
                "✓ Dockerfile estruturalmente válido "
                "(instrução FROM identificada)."
            )

        else:

            dockerfile_valido = False

            validacao_artefatos.append(
                "✗ Dockerfile sem instrução FROM identificada."
            )


# ---------------------------------------------------------
# 6. Docker Compose
# ---------------------------------------------------------

if docker_compose is not None:

    if not arquivo_nao_vazio(
        docker_compose
    ):

        docker_compose_valido = False

        validacao_artefatos.append(
            "✗ Docker Compose: arquivo vazio."
        )

    else:

        try:

            docker_compose_obj = yaml.safe_load(
                docker_compose
            )

            if not isinstance(
                docker_compose_obj,
                dict
            ):
                raise ValueError(
                    "estrutura YAML raiz não é um mapeamento"
                )

            if "services" not in docker_compose_obj:

                docker_compose_valido = False

                validacao_artefatos.append(
                    "✗ Docker Compose YAML válido, "
                    "mas sem seção 'services'."
                )

            else:

                docker_compose_valido = True

                validacao_artefatos.append(
                    "✓ Docker Compose YAML válido "
                    "e seção 'services' identificada."
                )

        except Exception as erro:

            docker_compose_valido = False

            validacao_artefatos.append(
                f"✗ Docker Compose YAML inválido: {erro}"
            )


# ---------------------------------------------------------
# 7. Program.cs
# ---------------------------------------------------------

if program_cs is not None:

    if not arquivo_nao_vazio(
        program_cs
    ):

        program_cs_valido = False

        validacao_artefatos.append(
            "✗ Program.cs: arquivo vazio."
        )

    else:

        indicadores_cs = (
            "WebApplication",
            "CreateBuilder",
            "builder.",
            "Build("
        )

        program_cs_valido = any(
            indicador.lower()
            in program_cs.lower()
            for indicador in indicadores_cs
        )

        if program_cs_valido:

            validacao_artefatos.append(
                "✓ Program.cs legível e com estrutura "
                "compatível com inicialização de aplicação .NET."
            )

        else:

            validacao_artefatos.append(
                "⚠ Program.cs legível, mas a estrutura "
                "esperada de inicialização .NET "
                "não foi identificada."
            )


# ---------------------------------------------------------
# 8. appsettings.json
# ---------------------------------------------------------

if appsettings is not None:

    if not arquivo_nao_vazio(
        appsettings
    ):

        appsettings_valido = False

        validacao_artefatos.append(
            "✗ appsettings.json: arquivo vazio."
        )

    else:

        try:

            appsettings_obj = json.loads(
                appsettings
            )

            if not isinstance(
                appsettings_obj,
                dict
            ):
                raise ValueError(
                    "raiz JSON não é um objeto"
                )

            appsettings_valido = True

            validacao_artefatos.append(
                "✓ appsettings.json válido."
            )

        except Exception as erro:

            appsettings_valido = False

            validacao_artefatos.append(
                f"✗ appsettings.json inválido: {erro}"
            )


# ---------------------------------------------------------
# 9. .editorconfig
# ---------------------------------------------------------

if editorconfig is not None:

    if not arquivo_nao_vazio(
        editorconfig
    ):

        editorconfig_valido = False

        validacao_artefatos.append(
            "✗ .editorconfig: arquivo vazio."
        )

    else:

        possui_secao = bool(
            re.search(
                r"^\s*\[[^\]]+\]\s*$",
                editorconfig,
                flags=re.MULTILINE
            )
        )

        editorconfig_valido = possui_secao

        if possui_secao:

            validacao_artefatos.append(
                "✓ .editorconfig estruturalmente válido "
                "(seção de configuração identificada)."
            )

        else:

            validacao_artefatos.append(
                "⚠ .editorconfig legível, "
                "mas nenhuma seção foi identificada."
            )


# ---------------------------------------------------------
# 10. Arquivos de ambiente
# ---------------------------------------------------------

env_example_valido = validar_env(
    env_example,
    ".env.example"
)

env_io_example_valido = validar_env(
    env_io_example,
    ".env.io.example"
)


# ---------------------------------------------------------
# 11. Scripts
# ---------------------------------------------------------

for script_nome in sorted(
    scripts_nomes
):

    conteudo_script = scripts.get(
        script_nome
    )

    if arquivo_nao_vazio(
        conteudo_script
    ):

        scripts_validos[
            script_nome
        ] = True

        extensao = Path(
            normalizar_nome(
                script_nome
            )
        ).suffix

        tipo_script = {
            ".sql": "SQL",
            ".ps1": "PowerShell",
            ".sh": "Shell",
            ".bat": "Batch",
            ".cmd": "Command"
        }.get(
            extensao,
            "script"
        )

        validacao_artefatos.append(
            f"✓ {script_nome}: "
            f"script {tipo_script} legível e não vazio."
        )

    else:

        scripts_validos[
            script_nome
        ] = False

        validacao_artefatos.append(
            f"✗ {script_nome}: arquivo vazio."
        )


# ---------------------------------------------------------
# Resultado da identificação
# ---------------------------------------------------------

print(
    "ARTEFATOS IDENTIFICADOS\n"
)


def imprimir_artefato(
    descricao,
    nome
):
    if nome:

        print(
            f"✓ {descricao}: {nome}"
        )

    else:

        print(
            f"? {descricao}: não fornecido"
        )


imprimir_artefato(
    "Pipeline",
    pipeline_nome
)

imprimir_artefato(
    ".csproj",
    csproj_nome
)

imprimir_artefato(
    "renovate.json",
    renovate_json_nome
)

imprimir_artefato(
    "RENOVATE_SETUP.md",
    renovate_setup_nome
)

imprimir_artefato(
    "Dockerfile",
    dockerfile_nome
)

imprimir_artefato(
    "Docker Compose",
    docker_compose_nome
)

imprimir_artefato(
    "Program.cs",
    program_cs_nome
)

imprimir_artefato(
    "appsettings.json",
    appsettings_nome
)

imprimir_artefato(
    ".editorconfig",
    editorconfig_nome
)

imprimir_artefato(
    ".env.example",
    env_example_nome
)

imprimir_artefato(
    ".env.io.example",
    env_io_example_nome
)


if scripts_nomes:

    print(
        f"✓ Scripts: "
        f"{len(scripts_nomes)} arquivo(s)"
    )

    for nome in sorted(
        scripts_nomes
    ):
        print(
            f"    - {nome}"
        )

else:

    print(
        "? Scripts: não fornecidos"
    )


# ---------------------------------------------------------
# Resultado da validação
# ---------------------------------------------------------

print(
    "\nVALIDAÇÃO\n"
)

if validacao_artefatos:

    for mensagem in validacao_artefatos:

        print(
            mensagem
        )

else:

    print(
        "Nenhum artefato sujeito a validação "
        "foi fornecido."
    )


# ---------------------------------------------------------
# Estado técnico da validação
# ---------------------------------------------------------

print(
    "\nESTADO DA VALIDAÇÃO\n"
)

estados_validacao = {
    "Pipeline": pipeline_valido,
    ".csproj": csproj_valido,
    "renovate.json": renovate_json_valido,
    "RENOVATE_SETUP.md": renovate_setup_valido,
    "Dockerfile": dockerfile_valido,
    "Docker Compose": docker_compose_valido,
    "Program.cs": program_cs_valido,
    "appsettings.json": appsettings_valido,
    ".editorconfig": editorconfig_valido,
    ".env.example": env_example_valido,
    ".env.io.example": env_io_example_valido
}


for artefato, estado in estados_validacao.items():

    if estado is True:
        simbolo = "✓"

    elif estado is False:
        simbolo = "✗"

    else:
        simbolo = "?"

    print(
        f"{simbolo} {artefato}"
    )
```

```python
# ---------------------------------------------------------
# PIPELINE MATURITY CLASSIFICATION
# ---------------------------------------------------------

def classificar_maturidade(e):

    if not e:

        return {
            "nivel": None,
            "classificacao": "Indeterminado",
            "motivo": (
                "Nenhuma evidência foi extraída."
            )
        }


    gitlab_ci = e.get(
        "GitLab CI como código"
    )

    restore = e.get(
        "`dotnet restore` no CI"
    )

    build = e.get(
        "Build Release no CI"
    )

    sonar = e.get(
        "SonarQube"
    )

    quality_gate_wait = e.get(
        "Espera pelo Quality Gate"
    )

    quality_gate_blocking = e.get(
        "Quality Gate bloqueante"
    )

    testes = e.get(
        "Testes automatizados"
    )

    deploy = e.get(
        "Deploy automatizado"
    )

    promocao = e.get(
        "Promoção automática"
    )

    rollback = e.get(
        "Rollback automatizado"
    )


    # -----------------------------------------------------
    # Evidência insuficiente
    # -----------------------------------------------------

    if all(
        valor is None
        for valor in (
            gitlab_ci,
            restore,
            build,
            sonar
        )
    ):

        return {
            "nivel": None,
            "classificacao": "Indeterminado",
            "motivo": (
                "O pipeline não está disponível ou não pôde "
                "ser validado."
            )
        }


    # -----------------------------------------------------
    # CI estruturada
    # -----------------------------------------------------

    ci_estruturada = (
        gitlab_ci is True
        and restore is True
        and build is True
    )

    qualidade = (
        sonar is True
        or quality_gate_wait is True
    )


    # -----------------------------------------------------
    # Nível 4
    # -----------------------------------------------------

    if (
        ci_estruturada
        and qualidade
        and testes is True
        and quality_gate_blocking is True
        and deploy is True
        and promocao is True
        and rollback is True
    ):

        return {
            "nivel": 4,
            "classificacao": "Avançado/Consolidado",
            "motivo": (
                "O pipeline apresenta CI estruturada, testes "
                "automatizados, controle de qualidade bloqueante, "
                "deploy, promoção e rollback automatizados."
            )
        }


    # -----------------------------------------------------
    # Nível 3
    # -----------------------------------------------------

    if (
        ci_estruturada
        and qualidade
    ):

        return {
            "nivel": 3,
            "classificacao": "Parcial/Emergente",
            "motivo": (
                "O pipeline apresenta integração contínua "
                "estruturada e controles de qualidade, mas "
                "a automação de ponta a ponta permanece incompleta."
            )
        }


    # -----------------------------------------------------
    # Nível 2
    # -----------------------------------------------------

    if (
        gitlab_ci is True
        and (
            restore is True
            or build is True
            or sonar is True
        )
    ):

        return {
            "nivel": 2,
            "classificacao": "Inicial/Ad hoc",
            "motivo": (
                "Existe pipeline versionado com automações "
                "pontuais, mas sem CI estruturada completa."
            )
        }


    # -----------------------------------------------------
    # Nível 1
    # -----------------------------------------------------

    if gitlab_ci is True:

        return {
            "nivel": 1,
            "classificacao": "Ausente/Tradicional",
            "motivo": (
                "O pipeline existe, mas não foram identificadas "
                "automações relevantes segundo os critérios definidos."
            )
        }


    return {
        "nivel": None,
        "classificacao": "Indeterminado",
        "motivo": (
            "As evidências disponíveis são insuficientes."
        )
    }


# ---------------------------------------------------------
# Executa a classificação
# ---------------------------------------------------------

resultado_maturidade = classificar_maturidade(
    evidencias
)


# ---------------------------------------------------------
# Resultado
# ---------------------------------------------------------

print("CLASSIFICAÇÃO DE MATURIDADE\n")

print(
    "Nível:",
    resultado_maturidade["nivel"]
)

print(
    "Classificação:",
    resultado_maturidade["classificacao"]
)

print(
    "Motivo:",
    resultado_maturidade["motivo"]
)
```

```python
# ---------------------------------------------------------
# ARTIFACT IDENTIFICATION AND VALIDATION
# ---------------------------------------------------------

if not arquivos:
    raise ValueError(
        "Nenhum arquivo foi carregado. "
        "Execute primeiro a etapa Artifact Upload."
    )


# ---------------------------------------------------------
# Reinicia variáveis de identificação
# ---------------------------------------------------------

pipeline = None
pipeline_nome = None

csproj = None
csproj_nome = None

renovate_json = None
renovate_json_nome = None

renovate_setup = None
renovate_setup_nome = None

dockerfile = None
dockerfile_nome = None

docker_compose = None
docker_compose_nome = None

program_cs = None
program_cs_nome = None

appsettings = None
appsettings_nome = None

editorconfig = None
editorconfig_nome = None

env_example = None
env_example_nome = None

env_io_example = None
env_io_example_nome = None

scripts = {}
scripts_nomes = []

validacao_artefatos = []


# ---------------------------------------------------------
# Estados formais de validação
#
# None  = artefato não fornecido / não validável
# True  = artefato fornecido e validado
# False = artefato fornecido, porém inválido
# ---------------------------------------------------------

pipeline_valido = None
csproj_valido = None
renovate_json_valido = None
renovate_setup_valido = None
dockerfile_valido = None
docker_compose_valido = None
program_cs_valido = None
appsettings_valido = None
editorconfig_valido = None
env_example_valido = None
env_io_example_valido = None

scripts_validos = {}


# ---------------------------------------------------------
# Normalização de nomes
# ---------------------------------------------------------

def nome_base(nome):
    return (
        nome
        .replace("\\", "/")
        .split("/")[-1]
        .strip()
    )


def normalizar_nome(nome):
    nome = nome_base(nome)

    # Remove sufixos adicionados pelo Colab:
    # arquivo (1).json -> arquivo.json
    # Dockerfile (2)   -> Dockerfile
    nome = re.sub(
        r"\s*\(\d+\)(?=\.|$)",
        "",
        nome
    )

    return nome.lower().strip()


# ---------------------------------------------------------
# Seleção de artefato único
# ---------------------------------------------------------

def selecionar_unico(
    candidatos,
    descricao
):
    if not candidatos:
        return None, None

    if len(candidatos) > 1:

        nomes = ", ".join(
            sorted(
                nome
                for nome, _ in candidatos
            )
        )

        raise ValueError(
            f"Mais de um arquivo foi identificado como "
            f"{descricao}: {nomes}. "
            "Forneça apenas um artefato desse tipo por versão."
        )

    return (
        candidatos[0][1],
        candidatos[0][0]
    )


# ---------------------------------------------------------
# Conversão segura para texto
# ---------------------------------------------------------

def como_texto(conteudo):
    if conteudo is None:
        return None

    if isinstance(
        conteudo,
        bytes
    ):
        return conteudo.decode(
            "utf-8-sig",
            errors="replace"
        )

    texto = str(
        conteudo
    )

    # Remove BOM caso o conteúdo já seja string.
    return texto.lstrip(
        "\ufeff"
    )


# ---------------------------------------------------------
# Candidatos
# ---------------------------------------------------------

candidatos_pipeline = []
candidatos_csproj = []
candidatos_renovate = []
candidatos_renovate_setup = []
candidatos_dockerfile = []
candidatos_docker_compose = []
candidatos_program = []
candidatos_appsettings = []
candidatos_editorconfig = []
candidatos_env_example = []
candidatos_env_io_example = []

extensoes_script = {
    ".sql",
    ".ps1",
    ".sh",
    ".bat",
    ".cmd"
}


# ---------------------------------------------------------
# Identificação dos arquivos
# ---------------------------------------------------------

for nome, conteudo in arquivos.items():

    base = nome_base(
        nome
    )

    normalizado = normalizar_nome(
        nome
    )


    # GitLab CI
    if normalizado in {
        ".gitlab-ci.yml",
        ".gitlab-ci.yaml",
        "gitlab-ci.yml",
        "gitlab-ci.yaml"
    }:

        candidatos_pipeline.append(
            (
                base,
                conteudo
            )
        )

        continue


    # Projeto .NET
    if normalizado.endswith(
        ".csproj"
    ):

        candidatos_csproj.append(
            (
                base,
                conteudo
            )
        )

        continue


    # Renovate
    if normalizado == "renovate.json":

        candidatos_renovate.append(
            (
                base,
                conteudo
            )
        )

        continue


    if normalizado == "renovate_setup.md":

        candidatos_renovate_setup.append(
            (
                base,
                conteudo
            )
        )

        continue


    # Dockerfile
    if normalizado == "dockerfile":

        candidatos_dockerfile.append(
            (
                base,
                conteudo
            )
        )

        continue


    # Docker Compose
    if normalizado in {
        "docker-compose.yml",
        "docker-compose.yaml",
        "compose.yml",
        "compose.yaml"
    }:

        candidatos_docker_compose.append(
            (
                base,
                conteudo
            )
        )

        continue


    # Program.cs
    if normalizado == "program.cs":

        candidatos_program.append(
            (
                base,
                conteudo
            )
        )

        continue


    # appsettings.json
    if normalizado == "appsettings.json":

        candidatos_appsettings.append(
            (
                base,
                conteudo
            )
        )

        continue


    # .editorconfig
    if normalizado == ".editorconfig":

        candidatos_editorconfig.append(
            (
                base,
                conteudo
            )
        )

        continue


    # .env.io.example
    if normalizado in {
        ".env.io.example",
        "env.io.example"
    }:

        candidatos_env_io_example.append(
            (
                base,
                conteudo
            )
        )

        continue


    # .env.example
    if normalizado in {
        ".env.example",
        "env.example"
    }:

        candidatos_env_example.append(
            (
                base,
                conteudo
            )
        )

        continue


    # Scripts
    extensao = Path(
        normalizado
    ).suffix

    if extensao in extensoes_script:

        scripts[
            base
        ] = como_texto(
            conteudo
        )

        scripts_nomes.append(
            base
        )


# ---------------------------------------------------------
# Seleção dos artefatos
# ---------------------------------------------------------

pipeline, pipeline_nome = selecionar_unico(
    candidatos_pipeline,
    "pipeline GitLab CI"
)

csproj, csproj_nome = selecionar_unico(
    candidatos_csproj,
    "arquivo .csproj"
)

renovate_json, renovate_json_nome = selecionar_unico(
    candidatos_renovate,
    "renovate.json"
)

renovate_setup, renovate_setup_nome = selecionar_unico(
    candidatos_renovate_setup,
    "RENOVATE_SETUP.md"
)

dockerfile, dockerfile_nome = selecionar_unico(
    candidatos_dockerfile,
    "Dockerfile"
)

docker_compose, docker_compose_nome = selecionar_unico(
    candidatos_docker_compose,
    "Docker Compose"
)

program_cs, program_cs_nome = selecionar_unico(
    candidatos_program,
    "Program.cs"
)

appsettings, appsettings_nome = selecionar_unico(
    candidatos_appsettings,
    "appsettings.json"
)

editorconfig, editorconfig_nome = selecionar_unico(
    candidatos_editorconfig,
    ".editorconfig"
)

env_example, env_example_nome = selecionar_unico(
    candidatos_env_example,
    ".env.example"
)

env_io_example, env_io_example_nome = selecionar_unico(
    candidatos_env_io_example,
    ".env.io.example"
)


# ---------------------------------------------------------
# Conversão para texto
# ---------------------------------------------------------

pipeline = como_texto(
    pipeline
)

csproj = como_texto(
    csproj
)

renovate_json = como_texto(
    renovate_json
)

renovate_setup = como_texto(
    renovate_setup
)

dockerfile = como_texto(
    dockerfile
)

docker_compose = como_texto(
    docker_compose
)

program_cs = como_texto(
    program_cs
)

appsettings = como_texto(
    appsettings
)

editorconfig = como_texto(
    editorconfig
)

env_example = como_texto(
    env_example
)

env_io_example = como_texto(
    env_io_example
)


# ---------------------------------------------------------
# Funções auxiliares de validação
# ---------------------------------------------------------

def arquivo_nao_vazio(
    conteudo
):
    return (
        conteudo is not None
        and bool(
            conteudo.strip()
        )
    )


def validar_env(
    conteudo,
    descricao
):
    if conteudo is None:
        return None

    if not arquivo_nao_vazio(
        conteudo
    ):
        validacao_artefatos.append(
            f"✗ {descricao}: arquivo vazio."
        )

        return False

    linhas_invalidas = []

    for numero, linha in enumerate(
        conteudo.splitlines(),
        start=1
    ):

        linha = linha.strip()

        if (
            not linha
            or linha.startswith("#")
        ):
            continue

        if "=" not in linha:
            linhas_invalidas.append(
                numero
            )

    if linhas_invalidas:

        validacao_artefatos.append(
            f"⚠ {descricao}: "
            f"{len(linhas_invalidas)} linha(s) "
            "sem formato CHAVE=VALOR."
        )

        # O arquivo continua legível e utilizável
        # para extração de evidências.
        return True

    validacao_artefatos.append(
        f"✓ {descricao} estruturalmente válido."
    )

    return True


# ---------------------------------------------------------
# 1. Pipeline YAML
# ---------------------------------------------------------

if pipeline is not None:

    if not arquivo_nao_vazio(
        pipeline
    ):

        pipeline_valido = False

        validacao_artefatos.append(
            "✗ Pipeline: arquivo vazio."
        )

    else:

        try:

            pipeline_yaml = yaml.safe_load(
                pipeline
            )

            if not isinstance(
                pipeline_yaml,
                dict
            ):
                raise ValueError(
                    "estrutura YAML raiz não é um mapeamento"
                )

            pipeline_valido = True

            validacao_artefatos.append(
                "✓ Pipeline YAML válido."
            )

        except Exception as erro:

            pipeline_valido = False

            validacao_artefatos.append(
                f"✗ Pipeline YAML inválido: {erro}"
            )


# ---------------------------------------------------------
# 2. .csproj XML
# ---------------------------------------------------------

if csproj is not None:

    if not arquivo_nao_vazio(
        csproj
    ):

        csproj_valido = False

        validacao_artefatos.append(
            "✗ .csproj: arquivo vazio."
        )

    else:

        try:

            ET.fromstring(
                csproj
            )

            csproj_valido = True

            validacao_artefatos.append(
                "✓ .csproj XML válido."
            )

        except ET.ParseError as erro:

            csproj_valido = False

            validacao_artefatos.append(
                f"✗ .csproj XML inválido: {erro}"
            )


# ---------------------------------------------------------
# 3. renovate.json
# ---------------------------------------------------------

if renovate_json is not None:

    if not arquivo_nao_vazio(
        renovate_json
    ):

        renovate_json_valido = False

        validacao_artefatos.append(
            "✗ renovate.json: arquivo vazio."
        )

    else:

        try:

            renovate_data_validacao = json.loads(
                renovate_json
            )

            if not isinstance(
                renovate_data_validacao,
                dict
            ):
                raise ValueError(
                    "raiz JSON não é um objeto"
                )

            renovate_json_valido = True

            validacao_artefatos.append(
                "✓ renovate.json válido."
            )

        except Exception as erro:

            renovate_json_valido = False

            validacao_artefatos.append(
                f"✗ renovate.json inválido: {erro}"
            )


# ---------------------------------------------------------
# 4. RENOVATE_SETUP.md
# ---------------------------------------------------------

if renovate_setup is not None:

    if arquivo_nao_vazio(
        renovate_setup
    ):

        renovate_setup_valido = True

        validacao_artefatos.append(
            "✓ RENOVATE_SETUP.md legível e não vazio."
        )

    else:

        renovate_setup_valido = False

        validacao_artefatos.append(
            "✗ RENOVATE_SETUP.md: arquivo vazio."
        )


# ---------------------------------------------------------
# 5. Dockerfile
# ---------------------------------------------------------

if dockerfile is not None:

    if not arquivo_nao_vazio(
        dockerfile
    ):

        dockerfile_valido = False

        validacao_artefatos.append(
            "✗ Dockerfile: arquivo vazio."
        )

    else:

        possui_from = bool(
            re.search(
                r"^\s*FROM\s+\S+",
                dockerfile,
                flags=(
                    re.IGNORECASE
                    | re.MULTILINE
                )
            )
        )

        if possui_from:

            dockerfile_valido = True

            validacao_artefatos.append(
                "✓ Dockerfile estruturalmente válido "
                "(instrução FROM identificada)."
            )

        else:

            dockerfile_valido = False

            validacao_artefatos.append(
                "✗ Dockerfile sem instrução FROM identificada."
            )


# ---------------------------------------------------------
# 6. Docker Compose
# ---------------------------------------------------------

if docker_compose is not None:

    if not arquivo_nao_vazio(
        docker_compose
    ):

        docker_compose_valido = False

        validacao_artefatos.append(
            "✗ Docker Compose: arquivo vazio."
        )

    else:

        try:

            docker_compose_obj = yaml.safe_load(
                docker_compose
            )

            if not isinstance(
                docker_compose_obj,
                dict
            ):
                raise ValueError(
                    "estrutura YAML raiz não é um mapeamento"
                )

            if "services" not in docker_compose_obj:

                docker_compose_valido = False

                validacao_artefatos.append(
                    "✗ Docker Compose YAML válido, "
                    "mas sem seção 'services'."
                )

            else:

                docker_compose_valido = True

                validacao_artefatos.append(
                    "✓ Docker Compose YAML válido "
                    "e seção 'services' identificada."
                )

        except Exception as erro:

            docker_compose_valido = False

            validacao_artefatos.append(
                f"✗ Docker Compose YAML inválido: {erro}"
            )


# ---------------------------------------------------------
# 7. Program.cs
# ---------------------------------------------------------

if program_cs is not None:

    if not arquivo_nao_vazio(
        program_cs
    ):

        program_cs_valido = False

        validacao_artefatos.append(
            "✗ Program.cs: arquivo vazio."
        )

    else:

        indicadores_cs = (
            "WebApplication",
            "CreateBuilder",
            "builder.",
            "Build("
        )

        program_cs_valido = any(
            indicador.lower()
            in program_cs.lower()
            for indicador in indicadores_cs
        )

        if program_cs_valido:

            validacao_artefatos.append(
                "✓ Program.cs legível e com estrutura "
                "compatível com inicialização de aplicação .NET."
            )

        else:

            validacao_artefatos.append(
                "⚠ Program.cs legível, mas a estrutura "
                "esperada de inicialização .NET "
                "não foi identificada."
            )


# ---------------------------------------------------------
# 8. appsettings.json
# ---------------------------------------------------------

if appsettings is not None:

    if not arquivo_nao_vazio(
        appsettings
    ):

        appsettings_valido = False

        validacao_artefatos.append(
            "✗ appsettings.json: arquivo vazio."
        )

    else:

        try:

            appsettings_obj = json.loads(
                appsettings
            )

            if not isinstance(
                appsettings_obj,
                dict
            ):
                raise ValueError(
                    "raiz JSON não é um objeto"
                )

            appsettings_valido = True

            validacao_artefatos.append(
                "✓ appsettings.json válido."
            )

        except Exception as erro:

            appsettings_valido = False

            validacao_artefatos.append(
                f"✗ appsettings.json inválido: {erro}"
            )


# ---------------------------------------------------------
# 9. .editorconfig
# ---------------------------------------------------------

if editorconfig is not None:

    if not arquivo_nao_vazio(
        editorconfig
    ):

        editorconfig_valido = False

        validacao_artefatos.append(
            "✗ .editorconfig: arquivo vazio."
        )

    else:

        possui_secao = bool(
            re.search(
                r"^\s*\[[^\]]+\]\s*$",
                editorconfig,
                flags=re.MULTILINE
            )
        )

        editorconfig_valido = possui_secao

        if possui_secao:

            validacao_artefatos.append(
                "✓ .editorconfig estruturalmente válido "
                "(seção de configuração identificada)."
            )

        else:

            validacao_artefatos.append(
                "⚠ .editorconfig legível, "
                "mas nenhuma seção foi identificada."
            )


# ---------------------------------------------------------
# 10. Arquivos de ambiente
# ---------------------------------------------------------

env_example_valido = validar_env(
    env_example,
    ".env.example"
)

env_io_example_valido = validar_env(
    env_io_example,
    ".env.io.example"
)


# ---------------------------------------------------------
# 11. Scripts
# ---------------------------------------------------------

for script_nome in sorted(
    scripts_nomes
):

    conteudo_script = scripts.get(
        script_nome
    )

    if arquivo_nao_vazio(
        conteudo_script
    ):

        scripts_validos[
            script_nome
        ] = True

        extensao = Path(
            normalizar_nome(
                script_nome
            )
        ).suffix

        tipo_script = {
            ".sql": "SQL",
            ".ps1": "PowerShell",
            ".sh": "Shell",
            ".bat": "Batch",
            ".cmd": "Command"
        }.get(
            extensao,
            "script"
        )

        validacao_artefatos.append(
            f"✓ {script_nome}: "
            f"script {tipo_script} legível e não vazio."
        )

    else:

        scripts_validos[
            script_nome
        ] = False

        validacao_artefatos.append(
            f"✗ {script_nome}: arquivo vazio."
        )


# ---------------------------------------------------------
# Resultado da identificação
# ---------------------------------------------------------

print(
    "ARTEFATOS IDENTIFICADOS\n"
)


def imprimir_artefato(
    descricao,
    nome
):
    if nome:

        print(
            f"✓ {descricao}: {nome}"
        )

    else:

        print(
            f"? {descricao}: não fornecido"
        )


imprimir_artefato(
    "Pipeline",
    pipeline_nome
)

imprimir_artefato(
    ".csproj",
    csproj_nome
)

imprimir_artefato(
    "renovate.json",
    renovate_json_nome
)

imprimir_artefato(
    "RENOVATE_SETUP.md",
    renovate_setup_nome
)

imprimir_artefato(
    "Dockerfile",
    dockerfile_nome
)

imprimir_artefato(
    "Docker Compose",
    docker_compose_nome
)

imprimir_artefato(
    "Program.cs",
    program_cs_nome
)

imprimir_artefato(
    "appsettings.json",
    appsettings_nome
)

imprimir_artefato(
    ".editorconfig",
    editorconfig_nome
)

imprimir_artefato(
    ".env.example",
    env_example_nome
)

imprimir_artefato(
    ".env.io.example",
    env_io_example_nome
)


if scripts_nomes:

    print(
        f"✓ Scripts: "
        f"{len(scripts_nomes)} arquivo(s)"
    )

    for nome in sorted(
        scripts_nomes
    ):
        print(
            f"    - {nome}"
        )

else:

    print(
        "? Scripts: não fornecidos"
    )


# ---------------------------------------------------------
# Resultado da validação
# ---------------------------------------------------------

print(
    "\nVALIDAÇÃO\n"
)

if validacao_artefatos:

    for mensagem in validacao_artefatos:

        print(
            mensagem
        )

else:

    print(
        "Nenhum artefato sujeito a validação "
        "foi fornecido."
    )


# ---------------------------------------------------------
# Estado técnico da validação
# ---------------------------------------------------------

print(
    "\nESTADO DA VALIDAÇÃO\n"
)

estados_validacao = {
    "Pipeline": pipeline_valido,
    ".csproj": csproj_valido,
    "renovate.json": renovate_json_valido,
    "RENOVATE_SETUP.md": renovate_setup_valido,
    "Dockerfile": dockerfile_valido,
    "Docker Compose": docker_compose_valido,
    "Program.cs": program_cs_valido,
    "appsettings.json": appsettings_valido,
    ".editorconfig": editorconfig_valido,
    ".env.example": env_example_valido,
    ".env.io.example": env_io_example_valido
}


for artefato, estado in estados_validacao.items():

    if estado is True:
        simbolo = "✓"

    elif estado is False:
        simbolo = "✗"

    else:
        simbolo = "?"

    print(
        f"{simbolo} {artefato}"
    )
```

```python
# ---------------------------------------------------------
# PIPELINE MATURITY CLASSIFICATION
# ---------------------------------------------------------

def classificar_maturidade(e):

    if not e:

        return {
            "nivel": None,
            "classificacao": "Indeterminado",
            "motivo": (
                "Nenhuma evidência foi extraída."
            )
        }


    gitlab_ci = e.get(
        "GitLab CI como código"
    )

    restore = e.get(
        "`dotnet restore` no CI"
    )

    build = e.get(
        "Build Release no CI"
    )

    sonar = e.get(
        "SonarQube"
    )

    quality_gate_wait = e.get(
        "Espera pelo Quality Gate"
    )

    quality_gate_blocking = e.get(
        "Quality Gate bloqueante"
    )

    testes = e.get(
        "Testes automatizados"
    )

    deploy = e.get(
        "Deploy automatizado"
    )

    promocao = e.get(
        "Promoção automática"
    )

    rollback = e.get(
        "Rollback automatizado"
    )


    # -----------------------------------------------------
    # Evidência insuficiente
    # -----------------------------------------------------

    if all(
        valor is None
        for valor in (
            gitlab_ci,
            restore,
            build,
            sonar
        )
    ):

        return {
            "nivel": None,
            "classificacao": "Indeterminado",
            "motivo": (
                "O pipeline não está disponível ou não pôde "
                "ser validado."
            )
        }


    # -----------------------------------------------------
    # CI estruturada
    # -----------------------------------------------------

    ci_estruturada = (
        gitlab_ci is True
        and restore is True
        and build is True
    )

    qualidade = (
        sonar is True
        or quality_gate_wait is True
    )


    # -----------------------------------------------------
    # Nível 4
    # -----------------------------------------------------

    if (
        ci_estruturada
        and qualidade
        and testes is True
        and quality_gate_blocking is True
        and deploy is True
        and promocao is True
        and rollback is True
    ):

        return {
            "nivel": 4,
            "classificacao": "Avançado/Consolidado",
            "motivo": (
                "O pipeline apresenta CI estruturada, testes "
                "automatizados, controle de qualidade bloqueante, "
                "deploy, promoção e rollback automatizados."
            )
        }


    # -----------------------------------------------------
    # Nível 3
    # -----------------------------------------------------

    if (
        ci_estruturada
        and qualidade
    ):

        return {
            "nivel": 3,
            "classificacao": "Parcial/Emergente",
            "motivo": (
                "O pipeline apresenta integração contínua "
                "estruturada e controles de qualidade, mas "
                "a automação de ponta a ponta permanece incompleta."
            )
        }


    # -----------------------------------------------------
    # Nível 2
    # -----------------------------------------------------

    if (
        gitlab_ci is True
        and (
            restore is True
            or build is True
            or sonar is True
        )
    ):

        return {
            "nivel": 2,
            "classificacao": "Inicial/Ad hoc",
            "motivo": (
                "Existe pipeline versionado com automações "
                "pontuais, mas sem CI estruturada completa."
            )
        }


    # -----------------------------------------------------
    # Nível 1
    # -----------------------------------------------------

    if gitlab_ci is True:

        return {
            "nivel": 1,
            "classificacao": "Ausente/Tradicional",
            "motivo": (
                "O pipeline existe, mas não foram identificadas "
                "automações relevantes segundo os critérios definidos."
            )
        }


    return {
        "nivel": None,
        "classificacao": "Indeterminado",
        "motivo": (
            "As evidências disponíveis são insuficientes."
        )
    }


# ---------------------------------------------------------
# Executa a classificação
# ---------------------------------------------------------

resultado_maturidade = classificar_maturidade(
    evidencias
)


# ---------------------------------------------------------
# Resultado
# ---------------------------------------------------------

print("CLASSIFICAÇÃO DE MATURIDADE\n")

print(
    "Nível:",
    resultado_maturidade["nivel"]
)

print(
    "Classificação:",
    resultado_maturidade["classificacao"]
)

print(
    "Motivo:",
    resultado_maturidade["motivo"]
)
```

```python
# ---------------------------------------------------------
# LONGITUDINAL PIPELINE TECHNICAL ANALYSIS — V1–V4
# ---------------------------------------------------------

versoes_esperadas = [
    "V1",
    "V2",
    "V3",
    "V4"
]


# ---------------------------------------------------------
# 1. Verifica se todas as análises foram armazenadas
# ---------------------------------------------------------

versoes_ausentes = [
    versao
    for versao in versoes_esperadas
    if versao not in resultados_versoes
]


if versoes_ausentes:

    raise ValueError(
        "As seguintes versões ainda não possuem resultados "
        "armazenados para análise longitudinal: "
        + ", ".join(versoes_ausentes)
    )


# ---------------------------------------------------------
# 2. Evidências selecionadas para comparação longitudinal
# ---------------------------------------------------------

evidencias_longitudinais = {

    "GitLab CI":
        "GitLab CI como código",

    "Restore no CI":
        "`dotnet restore` no CI",

    "Build Release":
        "Build Release no CI",

    "SonarQube":
        "SonarQube",

    "Espera pelo Quality Gate":
        "Espera pelo Quality Gate",

    "Quality Gate bloqueante":
        "Quality Gate bloqueante",

    "Cache Sonar":
        "Cache Sonar",

    "Cache NuGet":
        "Cache NuGet",

    "Regra para Merge Request":
        "Regra para Merge Request",

    "Regra para branch main":
        "Regra para branch main",

    "Testes automatizados":
        "Testes automatizados",

    "Renovate configurado":
        "Renovate configurado",

    "Renovate — NuGet":
        "Gerenciamento NuGet pelo Renovate",

    "Renovate — Dockerfile":
        "Gerenciamento Dockerfile pelo Renovate",

    "Vulnerability Alerts":
        "Vulnerability Alerts",

    "Lock File Maintenance":
        "Lock File Maintenance",

    "Job Renovate":
        "Job Renovate no pipeline",

    "Schedule Renovate":
        "Regra schedule para Renovate",

    "Execução manual Renovate":
        "Execução manual do Renovate",

    "Execução efetiva Renovate":
        "Execução efetiva do Renovate",

    "Dockerfile":
        "Dockerfile presente",

    "Docker multi-stage":
        "Docker multi-stage build",

    "Docker Compose":
        "Docker Compose presente",

    "Sentry em runtime":
        "Sentry integrado em runtime",

    "Health Checks":
        "Health Checks",

    "Matomo":
        "Matomo configurado",

    ".editorconfig":
        ".editorconfig presente",

    "Analyzer rules":
        "Regras de analyzer declaradas",

    "Security rules":
        "Regras de segurança declaradas",

    "Parametrização de ambientes":
        "Parametrização de ambientes",

    "Automação de ambiente local":
        "Automação de ambiente local",

    "Deploy automatizado":
        "Deploy automatizado",

    "Promoção automática":
        "Promoção automática",

    "Rollback automatizado":
        "Rollback automatizado"
}


# ---------------------------------------------------------
# 3. Conversão tri-state
# ---------------------------------------------------------

def simbolo_evidencia(valor):

    if valor is True:
        return "✓"

    if valor is False:
        return "—"

    return "?"


# ---------------------------------------------------------
# 4. Cria matriz longitudinal
# ---------------------------------------------------------

linhas = []

for descricao, chave in evidencias_longitudinais.items():

    linha = {
        "Evidência": descricao
    }

    for versao in versoes_esperadas:

        evidencias_versao = (
            resultados_versoes[
                versao
            ][
                "evidencias"
            ]
        )

        linha[versao] = simbolo_evidencia(
            evidencias_versao.get(
                chave
            )
        )

    linhas.append(
        linha
    )


df_longitudinal = pd.DataFrame(
    linhas
)


print(
    "MATRIZ LONGITUDINAL DE EVIDÊNCIAS — V1–V4\n"
)

display(
    df_longitudinal
)
```

```python
# ---------------------------------------------------------
# LONGITUDINAL MATURITY CLASSIFICATION
# ---------------------------------------------------------

linhas_maturidade = []


for versao in versoes_esperadas:

    resultado = resultados_versoes[
        versao
    ]

    classificacao = resultado[
        "classificacao"
    ]

    linhas_maturidade.append(
        {
            "Versão":
                versao,

            "Snapshot":
                resultado.get(
                    "snapshot"
                ),

            "Target Framework":
                resultado.get(
                    "target_framework"
                ),

            "SDK Pipeline":
                resultado.get(
                    "sdk_pipeline"
                ),

            "Nível":
                classificacao.get(
                    "nivel"
                ),

            "Classificação":
                classificacao.get(
                    "classificacao"
                ),

            "Motivo":
                classificacao.get(
                    "motivo"
                )
        }
    )


df_maturidade_longitudinal = pd.DataFrame(
    linhas_maturidade
)


print(
    "EVOLUÇÃO DA CLASSIFICAÇÃO DE MATURIDADE\n"
)

display(
    df_maturidade_longitudinal
)
```

```python
# ---------------------------------------------------------
# PREPARE LONGITUDINAL EVIDENCE
# ---------------------------------------------------------

evidencias_longitudinais_para_ia = {}


for versao in versoes_esperadas:

    resultado = resultados_versoes[
        versao
    ]

    evidencias_longitudinais_para_ia[
        versao
    ] = {

        "snapshot":
            resultado.get(
                "snapshot"
            ),

        "target_framework":
            resultado.get(
                "target_framework"
            ),

        "sdk_pipeline":
            resultado.get(
                "sdk_pipeline"
            ),

        "evidencias":
            resultado.get(
                "evidencias"
            ),

        "classificacao":
            resultado.get(
                "classificacao"
            )
    }


print(
    "✓ Evidências longitudinais estruturadas."
)
```

```python
# ---------------------------------------------------------
# LONGITUDINAL AI INTERPRETATION PROMPT
# ---------------------------------------------------------

prompt_longitudinal = f"""
Realize uma análise técnica longitudinal das práticas de
automação do pipeline nas versões V1, V2, V3 e V4.

A análise deve utilizar EXCLUSIVAMENTE os resultados das
análises individuais produzidas anteriormente pelo notebook.

Não reanalise os arquivos originais.

Não invente evidências adicionais.

Compare as versões considerando:

- permanência de práticas;
- introdução de novas capacidades;
- retirada de capacidades do pipeline;
- continuidade de configurações;
- diferenças entre configuração, orquestração e execução;
- evolução da classificação determinística;
- lacunas persistentes ao longo das versões.

Considere:

true  = evidência positiva;
false = ausência ou não evidência no artefato analisado;
null  = condição indeterminada.

Null não deve ser interpretado como ausência.

A configuração do Renovate não comprova sua execução efetiva.

A presença de um job Renovate demonstra orquestração, mas
não comprova execução bem-sucedida.

Dockerfile e Docker Compose não comprovam deploy automatizado.

Observabilidade não deve ser interpretada como automação
de entrega.

A ausência de comandos de deploy, promoção ou rollback nos
artefatos analisados não permite concluir que essas práticas
não ocorram fora do escopo analisado.

RESULTADOS DAS ANÁLISES INDIVIDUAIS:

{json.dumps(
    evidencias_longitudinais_para_ia,
    indent=2,
    ensure_ascii=False,
    default=str
)}

Produza a resposta em português nesta estrutura:

### Síntese longitudinal

### Evolução entre as versões
Analise V1 → V2 → V3 → V4.

### Práticas persistentes

### Capacidades introduzidas durante a evolução

### Mudanças ou descontinuidades de orquestração

### Lacunas persistentes

### Evolução da maturidade do pipeline

### Limitações da análise longitudinal

### Conclusão longitudinal

Use linguagem técnica, conservadora e adequada a um
trabalho acadêmico.
"""

print(
    "✓ Prompt longitudinal preparado."
)
```

```python
# ---------------------------------------------------------
# AI-ASSISTED LONGITUDINAL INTERPRETATION
# ---------------------------------------------------------

if not resultados_versoes:

    raise ValueError(
        "Os resultados individuais das versões "
        "não estão disponíveis."
    )


if not prompt_longitudinal:

    raise ValueError(
        "O prompt longitudinal não foi preparado."
    )


try:

    response_longitudinal = client.responses.create(

        model="gpt-5.6-terra",

        instructions=(
            "Atue como avaliador técnico de práticas DevOps. "
            "Realize uma interpretação longitudinal conservadora "
            "e rastreável. Utilize exclusivamente os resultados "
            "das análises individuais fornecidos. Não reclassifique "
            "arbitrariamente as versões e não invente evidências."
        ),

        input=prompt_longitudinal
    )


    interpretacao_longitudinal_ia = (
        response_longitudinal.output_text
    )


    print(
        "INTERPRETAÇÃO LONGITUDINAL "
        "ASSISTIDA POR IA — V1–V4\n"
    )

    print(
        interpretacao_longitudinal_ia
    )


except Exception as erro:

    interpretacao_longitudinal_ia = None

    print(
        "⚠ Não foi possível executar a "
        "interpretação longitudinal assistida por IA."
    )

    print(
        f"Detalhes: {erro}"
    )
```
